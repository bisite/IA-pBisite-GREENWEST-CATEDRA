```python
# -*- coding: utf-8 -*-
from db import engine
import pandas as pd
from sqlalchemy import Table, Column, MetaData, inspect
from sqlalchemy.dialects.mysql import DOUBLE, VARCHAR, TINYINT, INTEGER  # Tipos MySQL nativos

# Conexión

# ---------- Leer CSV ----------
ruta_parcela = "IFN_España/parcela.csv"
df_parcela = pd.read_csv(ruta_parcela)

# Columnas esperadas en el CSV
columnas_esperadas = [
    "idparcela", "coorx", "coory", "huso", "presencia",
    "tipsuelo1", "tipsuelo2", "tipsuelo3", "rocosidad",
    "radio", "superficie"
]
faltantes = [c for c in columnas_esperadas if c not in df_parcela.columns]
if faltantes:
    raise ValueError(f"Faltan columnas en el CSV: {faltantes}")

# ---------- Limpieza / normalización ----------
# Numéricos de punto flotante
for col_float in ["coorx", "coory", "radio", "superficie"]:
    df_parcela[col_float] = pd.to_numeric(df_parcela[col_float], errors="coerce")

# 'huso' es un entero UTM; lo guardaremos como INTEGER, pero permitimos NaN
df_parcela["huso"] = pd.to_numeric(df_parcela["huso"], errors="coerce").astype("Int64")

# idparcela a str y trim
df_parcela["idparcela"] = df_parcela["idparcela"].astype(str).str.strip()

# Campos discretos pequeños → enteros (nullable). Caben en TINYINT UNSIGNED (0..255)
for c in ["presencia", "tipsuelo1", "tipsuelo2", "tipsuelo3", "rocosidad"]:
    df_parcela[c] = pd.to_numeric(df_parcela[c], errors="coerce").astype("Int64")

# ---------- Definir/crear tabla ----------
metadata = MetaData()

parcelas = Table(
    "parcelas",
    metadata,
    Column("idparcela", VARCHAR(60), primary_key=True, nullable=False),     # PK NOT NULL
    Column("coorx", DOUBLE(asdecimal=False), nullable=True),
    Column("coory", DOUBLE(asdecimal=False), nullable=True),
    Column("huso", INTEGER(), nullable=True),                                # INTEGER NULL
    Column("presencia", TINYINT(), nullable=True),              # 0..255
    Column("tipsuelo1", TINYINT(), nullable=True),
    Column("tipsuelo2", TINYINT(), nullable=True),
    Column("tipsuelo3", TINYINT(), nullable=True),
    Column("rocosidad", TINYINT(), nullable=True),
    Column("radio", DOUBLE(asdecimal=False), nullable=True),
    Column("superficie", DOUBLE(asdecimal=False), nullable=True),
)

# Crear la tabla si no existe
metadata.create_all(engine, tables=[parcelas])

# ---------- Insertar datos ----------
# Mapear a dtypes compatibles antes de insertar
# Nota: como la tabla ya existe con los tipos correctos, usamos append
with engine.begin() as conn:
    df_parcela.to_sql(
        "parcelas",
        con=conn,
        if_exists="append",
        index=False,
        method="multi",
        chunksize=1000,
    )

print(f"✅ Insertadas {len(df_parcela):,} filas en 'parcelas'.")
```

```python
# -*- coding: utf-8 -*-
from db import engine
import pandas as pd
import numpy as np
from sqlalchemy import (
    create_engine, MetaData, Table, Column, ForeignKey,
    PrimaryKeyConstraint, inspect, text
)
from sqlalchemy.dialects.mysql import INTEGER, SMALLINT, TINYINT, VARCHAR, DOUBLE
from datetime import datetime

# -------------------------------------------------------------------------
# Rutas de entrada
ruta_ifn4 = "IFN_España/ifn4_parcela_inventario.csv"
ruta_ifn3 = "IFN_España/IFN3_campo/ifn3_campo_parcela.csv"
ruta_ifn2_especie = "IFN_España/ifn2_parcela_especie.csv"
ruta_ifn2_y_especie = "IFN_España/IFN2/ifn2_parcela_y_especie.csv"

# -------------------------------------------------------------------------
# Engine

# -------------------------------------------------------------------------
# 1) Leer IFN4 e IFN3
cols_needed = [
    'ano','nivel1','nivel2','merosiva','textura','matorg','modcomb',
    'disesp','comesp','fccarb','fcctot','idparcela','radio'
]

def leer_csv_seguro(path, usecols):
    cols_disponibles = pd.read_csv(path, nrows=0).columns.str.lower()
    usecols_presentes = [c for c in usecols if c in cols_disponibles]
    df = pd.read_csv(path, low_memory=False)
    df.columns = df.columns.str.lower()
    df = df[[c for c in usecols_presentes if c in df.columns]]
    for c in usecols:
        if c not in df.columns:
            df[c] = np.nan
    return df[usecols]

df_ifn4 = leer_csv_seguro(ruta_ifn4, cols_needed)
df_ifn4['idinventario'] = 4

df_ifn3 = leer_csv_seguro(ruta_ifn3, cols_needed)
df_ifn3['idinventario'] = 3

# -------------------------------------------------------------------------
# 2) IFN2: leer disesp/comesp e incorporar ano
df_ifn2_especie = pd.read_csv(ruta_ifn2_especie, low_memory=False)
df_ifn2_especie.columns = df_ifn2_especie.columns.str.lower()
df_ifn2_especie = df_ifn2_especie[[c for c in ['idparcela','disesp','comesp'] if c in df_ifn2_especie.columns]]

df_ifn2_y_especie = pd.read_csv(ruta_ifn2_y_especie, low_memory=False)
df_ifn2_y_especie.columns = df_ifn2_y_especie.columns.str.lower()
df_ifn2_y_especie = df_ifn2_y_especie[[c for c in ['idparcela','ano'] if c in df_ifn2_y_especie.columns]]

df_ifn2 = df_ifn2_especie.groupby('idparcela', as_index=False).first()
df_ifn2 = df_ifn2.merge(df_ifn2_y_especie, on='idparcela', how='left').rename(columns={})

for c in ['nivel1','nivel2','merosiva','textura','matorg','modcomb','fccarb','fcctot','radio']:
    df_ifn2[c] = np.nan
df_ifn2['idinventario'] = 2
df_ifn2 = df_ifn2[cols_needed + ['idinventario']].copy()

# -------------------------------------------------------------------------
# 3) Concatenar todo y normalizar tipos
df_all = pd.concat([df_ifn4, df_ifn3, df_ifn2], ignore_index=True)

df_all['idparcela']   = df_all['idparcela'].astype(str).str.strip()
df_all['idinventario'] = pd.to_numeric(df_all['idinventario'], errors='coerce').astype('Int64')
df_all['ano']          = pd.to_numeric(df_all['ano'], errors='coerce').astype('Int64')

for c in ['nivel1','nivel2','merosiva','textura','matorg','modcomb','disesp','comesp']:
    df_all[c] = pd.to_numeric(df_all[c], errors='coerce').astype('Int64')

for c in ['fccarb','fcctot','radio']:
    df_all[c] = pd.to_numeric(df_all[c], errors='coerce')

# -------------------------------------------------------------------------
# 3.b) Filtrar: mantener SOLO idparcela presentes en >1 inventario
inv_counts = (df_all.dropna(subset=['idparcela'])
                    .groupby('idparcela')['idinventario']
                    .nunique(dropna=True))
ids_multi_inv = set(inv_counts[inv_counts > 1].index)

before_rows = len(df_all)
df_all = df_all[df_all['idparcela'].isin(ids_multi_inv)].copy()
after_rows = len(df_all)
removed_rows = before_rows - after_rows

print(f"🔎 idparcela con >1 inventario: {len(ids_multi_inv):,}")
print(f"✂️ Filas eliminadas por criterio (>1 inventario): {removed_rows:,}")
print(f"✅ Filas restantes tras filtro: {after_rows:,}")

# -------------------------------------------------------------------------
# 4) Crear/ajustar tabla SQL con FK a parcelas.idparcela y PK compuesta
metadata = MetaData()
inspector = inspect(engine)
table_name = 'parcela_inventario'
parcelas_table_name = 'parcelas'

# Verificar que 'parcelas' existe
if not inspector.has_table(parcelas_table_name):
    raise SystemExit(f"La tabla '{parcelas_table_name}' no existe en la BBDD. Crea 'parcelas' antes.")

# Reflejar 'parcelas'
Table(parcelas_table_name, metadata, autoload_with=engine)

if not inspector.has_table(table_name):
    # Crear con el esquema exacto y PK compuesta (idparcela, idinventario)
    parcela_inventario_table = Table(
        table_name, metadata,
        Column('idparcela',   VARCHAR(60), ForeignKey(f"{parcelas_table_name}.idparcela"), nullable=False),
        Column('idinventario', INTEGER(), nullable=False),
        Column('ano',         SMALLINT(), nullable=True),
        Column('nivel1',      TINYINT(), nullable=True),
        Column('nivel2',      TINYINT(), nullable=True),
        Column('merosiva',    TINYINT(), nullable=True),
        Column('textura',     TINYINT(), nullable=True),
        Column('matorg',      TINYINT(), nullable=True),
        Column('modcomb',     TINYINT(), nullable=True),
        Column('disesp',      TINYINT(), nullable=True),
        Column('comesp',      TINYINT(), nullable=True),
        Column('fccarb',      DOUBLE(asdecimal=False), nullable=True),
        Column('fcctot',      DOUBLE(asdecimal=False), nullable=True),
        PrimaryKeyConstraint('idparcela','idinventario', name='pk_parcela_inventario'),
        mysql_charset='utf8mb4', mysql_collate='utf8mb4_unicode_ci'
    )
    metadata.create_all(engine, tables=[parcela_inventario_table])
    print(f"✅ Tabla '{table_name}' creada con PK compuesta y FK a {parcelas_table_name}.idparcela.")
else:
    print(f"ℹ️ La tabla '{table_name}' ya existe. Verificando/ajustando PK compuesta…")
    # Si ya existe, añadir PK compuesta si falta (elimina primero una PK previa si es distinta)
    with engine.begin() as conn:
        # ¿ya tiene PK compuesta correcta?
        has_pk = conn.execute(text("""
            SELECT COUNT(*) FROM information_schema.KEY_COLUMN_USAGE
            WHERE TABLE_SCHEMA = DATABASE()
              AND TABLE_NAME = :tbl
              AND CONSTRAINT_NAME = 'PRIMARY'
        """), {"tbl": table_name}).scalar()

        if has_pk:
            # Comprobar columnas de la PK
            pk_cols = conn.execute(text("""
                SELECT COLUMN_NAME, ORDINAL_POSITION
                FROM information_schema.KEY_COLUMN_USAGE
                WHERE TABLE_SCHEMA = DATABASE()
                  AND TABLE_NAME = :tbl
                  AND CONSTRAINT_NAME = 'PRIMARY'
                ORDER BY ORDINAL_POSITION
            """), {"tbl": table_name}).fetchall()
            pk_list = [r[0] for r in pk_cols]
            if pk_list != ['idparcela','idinventario']:
                # Cambiar PK
                conn.execute(text(f"ALTER TABLE {table_name} DROP PRIMARY KEY;"))
                conn.execute(text(f"ALTER TABLE {table_name} ADD PRIMARY KEY (idparcela, idinventario);"))
                print("🔧 PK reemplazada por (idparcela, idinventario).")
            else:
                print("✅ PK ya es (idparcela, idinventario).")
        else:
            conn.execute(text(f"ALTER TABLE {table_name} ADD PRIMARY KEY (idparcela, idinventario);"))
            print("✅ PK compuesta añadida (idparcela, idinventario).")

# -------------------------------------------------------------------------
# 5) Preparar datos para inserción (sin idparcela_inventario)
insert_cols = [
    'idparcela','idinventario','ano','nivel1','nivel2',
    'merosiva','textura','matorg','modcomb','disesp','comesp','fccarb','fcctot'
]
df_to_insert = df_all.copy()

# Coherencia de columnas + cast finales
df_to_insert['idparcela']    = df_to_insert['idparcela'].astype(str)
df_to_insert['idinventario'] = pd.to_numeric(df_to_insert['idinventario'], errors='coerce').astype('Int64')
df_to_insert['ano']          = pd.to_numeric(df_to_insert['ano'], errors='coerce').astype('Int64')

for c in ['nivel1','nivel2','merosiva','textura','matorg','modcomb','disesp','comesp']:
    df_to_insert[c] = pd.to_numeric(df_to_insert[c], errors='coerce').astype('Int64')

for c in ['fccarb','fcctot']:
    df_to_insert[c] = pd.to_numeric(df_to_insert[c], errors='coerce')

# ---  FILTRAR SOLO IDPARCELA EXISTENTES EN BD ---
with engine.connect() as conn:
    set_parcelas_db = set(pd.read_sql("SELECT idparcela FROM parcelas", conn)['idparcela'].astype(str).unique())

mask_in_parcelas = df_to_insert['idparcela'].astype(str).isin(set_parcelas_db)
df_to_insert = df_to_insert[mask_in_parcelas].copy()

# Enforce NOT NULL: idparcela, idinventario
before_nn = len(df_to_insert)
df_to_insert = df_to_insert.dropna(subset=['idparcela','idinventario'])
after_nn = len(df_to_insert)
if (before_nn - after_nn) > 0:
    print(f"⚠️ Filas eliminadas por violar NOT NULL ['idparcela','idinventario']: {before_nn - after_nn:,}")

# Quitar duplicados por PK (idparcela, idinventario) para no violar la PK
before_dups = len(df_to_insert)
df_to_insert = df_to_insert.drop_duplicates(subset=['idparcela','idinventario'])
after_dups = len(df_to_insert)
if (before_dups - after_dups) > 0:
    print(f"⚠️ Filas duplicadas eliminadas por PK: {before_dups - after_dups:,}")

# Dejar solo columnas del esquema
df_ok = df_to_insert[insert_cols].copy()

print(f"✅ Filas a insertar (tras filtros y deduplicación PK): {len(df_ok):,}")

# --- INSERTAR ---
chunksize = 1000
n_inserted = 0
with engine.begin() as conn:
    for start in range(0, len(df_ok), chunksize):
        chunk = df_ok.iloc[start:start+chunksize]
        # Asegúrate de que el orden de columnas coincide con la tabla
        chunk.to_sql(table_name, con=conn, if_exists='append', index=False, method='multi')
        n_inserted += len(chunk)

print(f"\n✅ Inserción completada. Filas insertadas: {n_inserted:,}.")

# -------------------------------------------------------------------------
# 6) Estadísticas
counts_by_inv = df_all.groupby('idinventario').size().to_dict()
unique_idparcela_by_inv = df_all.groupby('idinventario')['idparcela'].nunique().to_dict()
total_rows = len(df_all)
total_unique_pk = df_all.drop_duplicates(['idparcela','idinventario']).shape[0]
total_unique_idparcela = df_all['idparcela'].nunique()

print("\n--- Estadísticas (tras filtro multinventario) ---")
for inv, cnt in counts_by_inv.items():
    nids = unique_idparcela_by_inv.get(inv, 0)
    print(f"Inventario {inv}: filas={cnt:,} | idparcela únicos={nids:,}")
print(f"Total filas: {total_rows:,}")
print(f"Total PK únicas (idparcela,idinventario): {total_unique_pk:,}")
print(f"Total idparcela únicos: {total_unique_idparcela:,}")

# -------------------------------------------------------------------------
# 7) Cobertura respecto a 'parcelas'
try:
    with engine.connect() as conn:
        df_parcelas_db = pd.read_sql_table('parcelas', con=conn)
    set_parcelas_db = set(df_parcelas_db['idparcela'].astype(str).unique())
    set_inventario = set(df_all['idparcela'].astype(str).unique())
    n_in_parcelas = len(set_inventario & set_parcelas_db)
    pct_cov = 100.0 * n_in_parcelas / len(set_parcelas_db) if len(set_parcelas_db) > 0 else np.nan
    print("\n--- Cobertura ---")
    print(f"Parcelas en BD: {len(set_parcelas_db):,}")
    print(f"Coinciden: {n_in_parcelas:,} ({pct_cov:.2f}% de 'parcelas')")
except Exception as e:
    print("No se pudo leer la tabla 'parcelas'. Error:", e)
```

```python
# -*- coding: utf-8 -*-
from db import engine
from sqlalchemy import text
from datetime import datetime

ALTER_PRESENCIA_TO_TINYINT = True   # False para no alterar el tipo
DRY_RUN = False                     # True: imprime SQL, no ejecuta


def run(conn, sql):
    if DRY_RUN:
        print(sql.strip().rstrip(";") + ";")
    else:
        conn.execute(text(sql))

with engine.begin() as conn:
    # 0) (opcional) asegurar tipo de 'presencia'
    if ALTER_PRESENCIA_TO_TINYINT:
        run(conn, """
            ALTER TABLE parcelas
            MODIFY COLUMN presencia TINYINT UNSIGNED NULL
        """)

    # 1) Recalcular 'presencia' SOLO desde parcela_inventario (nombres actuales)
    run(conn, """
        UPDATE parcelas p
        LEFT JOIN (
            SELECT
              pi.idparcela,
              BIT_OR(CASE pi.idinventario
                        WHEN 2 THEN 1
                        WHEN 3 THEN 2
                        WHEN 4 THEN 4
                        ELSE 0
                     END) AS mask_bits
            FROM parcela_inventario pi
            GROUP BY pi.idparcela
        ) x ON x.idparcela = p.idparcela
        SET p.presencia = CASE x.mask_bits
            WHEN 7 THEN 1      -- 2&3&4
            WHEN 5 THEN 2      -- 2&4 (sin 3)
            WHEN 6 THEN 3      -- 3&4 (sin 2)
            ELSE NULL
        END
    """)

    # 1b) Info rápida
    dist = conn.execute(text("""
        SELECT
          CASE WHEN presencia IS NULL THEN 'NULL' ELSE CAST(presencia AS CHAR) END AS categoria,
          COUNT(*) AS n
        FROM parcelas
        GROUP BY categoria
        ORDER BY
          CASE categoria WHEN '1' THEN 1 WHEN '2' THEN 2 WHEN '3' THEN 3 WHEN 'NULL' THEN 4 ELSE 5 END
    """)).mappings().all()
    print("📊 Distribución de parcelas.presencia tras recálculo:")
    for r in dist:
        print(f"  {r['categoria']}: {r['n']}")

    # 2) Identificar candidatas a borrado (no 1/2/3)
    run(conn, "DROP TEMPORARY TABLE IF EXISTS tmp_parcelas_borrar")
    run(conn, """
        CREATE TEMPORARY TABLE tmp_parcelas_borrar AS
        SELECT idparcela
        FROM parcelas
        WHERE presencia NOT IN (1,2,3) OR presencia IS NULL
    """)
    n_candidatas = conn.execute(text("SELECT COUNT(*) FROM tmp_parcelas_borrar")).scalar()
    n_hijos = conn.execute(text("""
        SELECT COUNT(*) FROM parcela_inventario pi
        JOIN tmp_parcelas_borrar t USING (idparcela)
    """)).scalar()
    print(f"\n🎯 Parcelas a eliminar (presencia ∉ {{1,2,3}} o NULL): {n_candidatas} | filas hijas en PI: {n_hijos}")

    if n_candidatas > 0:
        # 3) Backups
        ts = datetime.utcnow().strftime("%Y%m%d%H%M%S")
        bkp_parc = f"bkp_parcelas_del_{ts}"
        bkp_pi   = f"bkp_parcela_inventario_del_{ts}"
        # crear tablas de backup (esquema simplificado)
        run(conn, f"CREATE TABLE `{bkp_parc}` (idparcela VARCHAR(60) PRIMARY KEY)")
        run(conn, f"INSERT INTO `{bkp_parc}` (idparcela) SELECT idparcela FROM tmp_parcelas_borrar")
        # backup de hijos PI (todas las columnas típicas)
        run(conn, f"""
            CREATE TABLE `{bkp_pi}` AS
            SELECT pi.*
            FROM parcela_inventario pi
            JOIN tmp_parcelas_borrar t USING (idparcela)
        """)
        print(f"💾 Backups creados: {bkp_parc} (IDs) y {bkp_pi} (filas PI)")

        # 4) Borrar primero HIJOS, luego PADRES (respeta la FK)
        run(conn, """
            DELETE pi
            FROM parcela_inventario pi
            JOIN tmp_parcelas_borrar t USING (idparcela)
        """)
        run(conn, """
            DELETE p
            FROM parcelas p
            JOIN tmp_parcelas_borrar t USING (idparcela)
        """)
        if not DRY_RUN:
            restantes = conn.execute(text("SELECT COUNT(*) FROM parcelas")).scalar()
            print(f"🗑️ Borrado realizado. Parcelas restantes: {restantes}")
        else:
            print("🗑️ DRY_RUN=True → se habría ejecutado el borrado de PI y luego de parcelas.")

    # 5) Resumen final
    if not DRY_RUN:
        dist2 = conn.execute(text("""
            SELECT
              CASE WHEN presencia IS NULL THEN 'NULL' ELSE CAST(presencia AS CHAR) END AS categoria,
              COUNT(*) AS n
            FROM parcelas
            GROUP BY categoria
            ORDER BY
              CASE categoria WHEN '1' THEN 1 WHEN '2' THEN 2 WHEN '3' THEN 3 WHEN 'NULL' THEN 4 ELSE 5 END
        """)).mappings().all()
        print("\n📊 Distribución de parcelas.presencia tras borrado:")
        for r in dist2:
            print(f"  {r['categoria']}: {r['n']}")
```

```python
# -*- coding: utf-8 -*-
from db import engine
import pandas as pd
from sqlalchemy import MetaData, Table, Column, inspect, text
from sqlalchemy.dialects.mysql import SMALLINT, TINYINT, TEXT  # Tipos MySQL nativos
from datetime import datetime

# ---------- CONFIG ----------
INPUT_CSV = "especies_to_load.csv"
CHUNKSIZE = 2000
RANGO_MAX_TRES_DIGITOS = 999  # ≤ 3 dígitos para idespecie/idgrupo

# ---------- LECTURA CSV ----------
print("📖 Leyendo CSV:", INPUT_CSV)
df_raw = pd.read_csv(INPUT_CSV, dtype=object)

# ---------- MAPEO DE COLUMNAS (acepta variantes de nombre) ----------
colmap = {
    "idespecie":   ["idespecie", "especie", "id_especie"],
    "nombre":      ["nombre", "especie_nombre"],
    "sinonimia":   ["sinonimia", "sinónimo", "sinonimo"],
    "tipo_especie":["tipo_especie", "conif_fronf", "tipo"],
    "idgrupo":     ["idgrupo", "grupo", "especie_grupo"],
    "nombregrupo": ["nombregrupo", "nombre_grupo", "grupo_nombre"],
}
df = pd.DataFrame()
for final_name, candidates in colmap.items():
    for c in candidates:
        if c in df_raw.columns:
            df[final_name] = df_raw[c]
            break
    if final_name not in df.columns:
        df[final_name] = pd.NA

# ---------- NORMALIZACIÓN / TIPOS ----------
# ids → SMALLINT (Int64 para permitir NA) y validar 0..999
df["idespecie"] = pd.to_numeric(df["idespecie"], errors="coerce").astype("Int64")
df["idgrupo"]   = pd.to_numeric(df["idgrupo"],   errors="coerce").astype("Int64")

def a_binario_01(x):
    if x is None or (isinstance(x, float) and pd.isna(x)) or pd.isna(x):
        return pd.NA
    s = str(x).strip().lower()
    if s in {"1","true","t","sí","si","s","y","yes","conif","conifera","conífera"}:
        return 1
    if s in {"0","false","f","no","n","frond","frondosa","frondoso"}:
        return 0
    try:
        v = int(float(s))
        return v if v in (0,1) else pd.NA
    except:
        return pd.NA

df["tipo_especie"] = df["tipo_especie"].apply(a_binario_01).astype("Int64")

# TEXT: NaN → None
for c in ["nombre", "sinonimia", "nombregrupo"]:
    df[c] = df[c].where(pd.notnull(df[c]), None)

# Descarta filas sin PK (idespecie)
before = len(df)
df = df.dropna(subset=["idespecie"])
print(f"🧹 Filas totales: {before:,} | tras descartar idespecie NULL: {len(df):,}")

# Validación de rangos 0..999
bad_idesp = df["idespecie"].dropna().astype(int).between(0, RANGO_MAX_TRES_DIGITOS) == False
bad_idgrp = df["idgrupo"].dropna().astype(int).between(0, RANGO_MAX_TRES_DIGITOS) == False
n_bad_idesp = int(bad_idesp.sum())
n_bad_idgrp = int(bad_idgrp.sum())
if n_bad_idesp or n_bad_idgrp:
    print(f"⚠️ Fuera de rango (0..{RANGO_MAX_TRES_DIGITOS}) → idespecie={n_bad_idesp}, idgrupo={n_bad_idgrp}. Se filtrarán.")
    df = df[
        (df["idespecie"].isna() | df["idespecie"].astype("Int64").between(0, RANGO_MAX_TRES_DIGITOS))
        &
        (df["idgrupo"].isna()   | df["idgrupo"].astype("Int64").between(0, RANGO_MAX_TRES_DIGITOS))
    ]
    print(f"🔧 Filas tras filtrar por rango: {len(df):,}")

# ---------- CONSTRUCCIÓN DE TABLAS NORMALIZADAS ----------
metadata = MetaData()
inspector = inspect(engine)

def backup_if_exists(conn, table_name):
    if inspector.has_table(table_name):
        ts = datetime.now().strftime("%Y%m%d_%H%M%S")
        backup_name = f"{table_name}_backup_{ts}"
        print(f"🔁 Existe '{table_name}'. Backup → '{backup_name}'…")
        conn.execute(text(f"RENAME TABLE `{table_name}` TO `{backup_name}`;"))
        print(f"✅ Backup creado: {backup_name}")

with engine.begin() as conn:
    # Backup de tablas si existen
    backup_if_exists(conn, "especies")
    backup_if_exists(conn, "grupos")

# Definir esquemas:
grupos_table = Table(
    "grupos", metadata,
    Column("idgrupo", SMALLINT(), primary_key=True, nullable=False),
    Column("nombregrupo", TEXT(), nullable=True),
    mysql_charset="utf8mb4",
)
especies_table = Table(
    "especies", metadata,
    Column("idespecie", SMALLINT(), primary_key=True, nullable=False),
    Column("nombre", TEXT(), nullable=True),
    Column("sinonimia", TEXT(), nullable=True),
    Column("tipo_especie", TINYINT(), nullable=True),
    Column("grupo", SMALLINT(), nullable=True),  # FK → grupos.idgrupo
    mysql_charset="utf8mb4",
)

metadata.create_all(engine, tables=[grupos_table, especies_table])
print("✅ Tablas 'grupos' y 'especies' creadas.")

# ---------- PREPARAR DATOS A INSERTAR ----------
# Tabla grupos: deduplicar por idgrupo, elige el primer nombregrupo no nulo por idgrupo
grp = (df[["idgrupo", "nombregrupo"]]
       .dropna(subset=["idgrupo"])
       .drop_duplicates(subset=["idgrupo"], keep="first")
       .rename(columns={"idgrupo":"idgrupo", "nombregrupo":"nombregrupo"}))

grp["idgrupo"] = grp["idgrupo"].astype("Int64")

# Tabla especies: mapea grupo = idgrupo; ya sin nombregrupo
esp = df[["idespecie", "nombre", "sinonimia", "tipo_especie", "idgrupo"]].copy()
esp = esp.rename(columns={"idgrupo":"grupo"})
esp["idespecie"]    = esp["idespecie"].astype("Int64")
esp["grupo"]        = esp["grupo"].astype("Int64")
esp["tipo_especie"] = esp["tipo_especie"].astype("Int64")

# ---------- INSERCIÓN ----------
print("⬆️ Insertando 'grupos'…")
if not grp.empty:
    grp.to_sql("grupos", con=engine, if_exists="append", index=False, method="multi", chunksize=CHUNKSIZE)
print(f"✅ 'grupos' insertado: {len(grp):,} filas.")

print("⬆️ Insertando 'especies'…")
if not esp.empty:
    esp[["idespecie","nombre","sinonimia","tipo_especie","grupo"]].to_sql(
        "especies", con=engine, if_exists="append", index=False, method="multi", chunksize=CHUNKSIZE
    )
print(f"✅ 'especies' insertado: {len(esp):,} filas.")

# ---------- CLAVE FORÁNEA (si no existe) ----------
with engine.begin() as conn:
    # Asegura tipo SMALLINT en 'grupo'
    conn.execute(text("ALTER TABLE especies MODIFY COLUMN grupo SMALLINT NULL;"))
    # Crea FK si no existe ya (consultamos information_schema)
    fk_exists = conn.execute(text("""
        SELECT COUNT(*) FROM information_schema.TABLE_CONSTRAINTS
        WHERE CONSTRAINT_SCHEMA = DATABASE()
          AND TABLE_NAME = 'especies'
          AND CONSTRAINT_TYPE = 'FOREIGN KEY'
          AND CONSTRAINT_NAME = 'fk_especies_grupo';
    """)).scalar()
    if not fk_exists:
        try:
            conn.execute(text("""
                ALTER TABLE especies
                  ADD CONSTRAINT fk_especies_grupo
                  FOREIGN KEY (grupo) REFERENCES grupos(idgrupo)
                  ON UPDATE CASCADE
                  ON DELETE SET NULL;
            """))
            print("🔗 FK creada: especies(grupo) → grupos(idgrupo)")
        except Exception as e:
            print("⚠️ No se pudo crear la FK (posiblemente ya existe o tipos incompatibles):", e)

# ---------- VERIFICACIÓN ----------
with engine.connect() as conn:
    n_grupos    = conn.execute(text("SELECT COUNT(*) FROM grupos")).scalar()
    n_especies  = conn.execute(text("SELECT COUNT(*) FROM especies")).scalar()
    head_grupos = pd.read_sql("SELECT idgrupo, nombregrupo FROM grupos ORDER BY idgrupo LIMIT 10", conn)
    head_esp    = pd.read_sql("SELECT idespecie, nombre, sinonimia, tipo_especie, grupo FROM especies ORDER BY idespecie LIMIT 10", conn)
    # Chequeo referencial: especies con grupo que no existe en grupos (debería ser 0)
    orfanas = pd.read_sql("""
        SELECT e.idespecie, e.grupo
        FROM especies e
        LEFT JOIN grupos g ON e.grupo = g.idgrupo
        WHERE e.grupo IS NOT NULL AND g.idgrupo IS NULL
        LIMIT 20
    """, conn)

print(f"📊 Totales → grupos: {int(n_grupos):,} | especies: {int(n_especies):,}")
print("\n🔎 Muestra 'grupos':")
display(head_grupos)
print("\n🔎 Muestra 'especies':")
display(head_esp)
if len(orfanas) == 0:
    print("\n✅ Integridad referencial OK (sin grupos huérfanos).")
else:
    print("\n⚠️ Hay especies con grupo sin correspondencia en 'grupos' (primeras filas):")
    display(orfanas)
```

```python
# -*- coding: utf-8 -*-
from db import engine
import pandas as pd
import numpy as np
import time
import re
from datetime import datetime
from sqlalchemy import (
    create_engine, inspect, text, MetaData, Table, Column,
    ForeignKey, ForeignKeyConstraint, PrimaryKeyConstraint
)
from sqlalchemy.dialects.mysql import VARCHAR, INTEGER, SMALLINT, TINYINT

# ---------- CONFIG ----------
ruta_ifn4_especie = "IFN_España/ifn4_parcela_especie.csv"
ruta_ifn3_especie = "IFN_España/ifn3_parcela_especie.csv"
ruta_ifn2_especie = "IFN_España/ifn2_parcela_especie.csv"

cols_ifn34 = ['idparcela','especie','ocupa','estado','fpmasa','tratmasa','orgmasa1']
cols_ifn2   = ['idparcela','especie','ocupa','estado','masa','origen']

TABLE_NAME = 'parcela_inventario_especie'
ESPECIES_TABLE = 'especies'
AUTO_CREATE_MISSING = True
OUT_MISSING  = "missing_especies_for_parcela_inventario_especie.csv"
OUT_REJECTED = "rejected_parcela_inventario_especie_missing_fk.csv"

# ---------- UTIL ----------
def safe_int_from(val):
    if pd.isna(val): 
        return None
    try:
        return int(float(val))
    except Exception:
        m = re.search(r"-?\d+", str(val))
        if m:
            try:
                return int(m.group(0))
            except:
                return None
    return None

def read_minimal(path, want_cols, dtype_map=None):
    head = pd.read_csv(path, nrows=0)
    present = [c for c in want_cols if c in head.columns]
    df = pd.read_csv(path, usecols=present, dtype=dtype_map or {c: str for c in present})
    for c in want_cols:
        if c not in df.columns:
            df[c] = np.nan
    return df

# ---------- LECTURA ----------
dtype_str = {c: str for c in set(cols_ifn34 + cols_ifn2)}
t0 = time.time()
df4 = read_minimal(ruta_ifn4_especie, cols_ifn34, dtype_map=dtype_str); df4['idinventario'] = 4; df4['masa'] = np.nan; df4['origen'] = np.nan
df3 = read_minimal(ruta_ifn3_especie, cols_ifn34, dtype_map=dtype_str); df3['idinventario'] = 3; df3['masa'] = np.nan; df3['origen'] = np.nan
df2 = read_minimal(ruta_ifn2_especie, cols_ifn2,   dtype_map=dtype_str); df2['idinventario'] = 2; df2['fpmasa'] = np.nan; df2['tratmasa'] = np.nan; df2['orgmasa1'] = np.nan
print(f"Lectura CSVs completada en {time.time()-t0:.1f}s")

final_cols = ['idparcela','especie','ocupa','estado','fpmasa','tratmasa','orgmasa1','masa','origen','idinventario']
df_all = pd.concat([df4[final_cols], df3[final_cols], df2[final_cols]], ignore_index=True, sort=False)

# ---------- NORMALIZACIÓN ----------
df_all['idparcela'] = df_all['idparcela'].astype(str).str.strip()

df_all['idespecie'] = df_all['especie'].apply(safe_int_from).astype('Int64')
df_all = df_all.drop(columns=['especie'])

df_all['idinventario'] = pd.to_numeric(df_all['idinventario'], errors='coerce').astype('Int64')

# categóricas a enteros pequeños (nullable)
for c in ['ocupa','estado','fpmasa','tratmasa','orgmasa1','masa','origen']:
    df_all[c] = pd.to_numeric(df_all[c], errors='coerce').astype('Int64')

# Eliminar filas con PK nula
df_all = df_all.dropna(subset=['idparcela','idinventario','idespecie'])

# Deduplicar por PK compuesta
n_before = len(df_all)
df_all = df_all.drop_duplicates(subset=['idparcela','idinventario','idespecie']).reset_index(drop=True)
if len(df_all) != n_before:
    print(f"Eliminados {n_before - len(df_all):,} duplicados por PK (idparcela, idinventario, idespecie)")

# ---------- CONEXIÓN & SETS FK ----------
inspector = inspect(engine)

with engine.connect() as conn:
    # FK simple -> especies(idespecie)
    if not inspector.has_table(ESPECIES_TABLE):
        raise SystemExit(f"No existe la tabla '{ESPECIES_TABLE}'. Crea 'especies' primero.")
    especies_db_set = set(pd.read_sql(f"SELECT idespecie FROM {ESPECIES_TABLE}", conn)['idespecie'].dropna().astype(int).unique())

    # FK compuesta -> parcela_inventario(idparcela, idinventario)
    if not inspector.has_table('parcela_inventario'):
        raise SystemExit("No existe la tabla 'parcela_inventario'. Créala primero con PK (idparcela, idinventario).")
    df_pi = pd.read_sql("SELECT idparcela, idinventario FROM parcela_inventario", conn)
    fk_pi_tuples = set(map(tuple, df_pi.astype({'idparcela':str,'idinventario':int}).values))

print(f"Especies registradas: {len(especies_db_set):,} | Pares (idparcela,idinventario) en PI: {len(fk_pi_tuples):,}")

# ---------- FILTRADO POR FKs ----------
required_not_null = ['idparcela','idinventario','idespecie']
df_all_nn = df_all.dropna(subset=required_not_null).copy()

mask_missing_especie = ~df_all_nn['idespecie'].astype(int).isin(especies_db_set)
mask_missing_pair    = ~df_all_nn.apply(lambda r: (str(r['idparcela']), int(r['idinventario'])) in fk_pi_tuples, axis=1)

df_rejected = df_all_nn[mask_missing_especie | mask_missing_pair].copy()
df_ok       = df_all_nn[~(mask_missing_especie | mask_missing_pair)].copy()

print(f"Filas totales: {len(df_all):,} | A insertar: {len(df_ok):,} | Rechazadas por FK: {len(df_rejected):,}")

# Guardar rechazadas y (opcional) crear placeholders de especies
if len(df_rejected):
    df_rejected.to_csv(OUT_REJECTED, index=False, encoding='utf-8')
    print(f"Rechazadas guardadas en '{OUT_REJECTED}'")

    missing_especies = sorted(df_rejected.loc[mask_missing_especie.reindex(df_rejected.index, fill_value=False), 'idespecie'].dropna().astype(int).unique())
    if missing_especies:
        pd.DataFrame({"idespecie_missing": missing_especies}).to_csv(OUT_MISSING, index=False, encoding='utf-8')
        print(f"Especies faltantes listadas en {OUT_MISSING} ({len(missing_especies):,}).")

        if AUTO_CREATE_MISSING:
            print("Creando placeholders en 'especies' (solo PK)…")
            with engine.begin() as conn:
                for code in missing_especies:
                    conn.execute(
                        text("INSERT INTO especies (idespecie) VALUES (:c) ON DUPLICATE KEY UPDATE idespecie=idespecie"),
                        {"c": int(code)}
                    )
            # recomputar tras placeholders
            with engine.connect() as conn:
                especies_db_set = set(pd.read_sql(f"SELECT idespecie FROM {ESPECIES_TABLE}", conn)['idespecie'].dropna().astype(int).unique())
            mask_missing_especie = ~df_all_nn['idespecie'].astype(int).isin(especies_db_set)
            df_rejected = df_all_nn[mask_missing_especie | mask_missing_pair].copy()
            df_ok       = df_all_nn[~(mask_missing_especie | mask_missing_pair)].copy()
            if len(df_rejected):
                df_rejected.to_csv(OUT_REJECTED, index=False, encoding='utf-8')
                print(f"Tras placeholders, aún rechazadas: {len(df_rejected):,} (ver '{OUT_REJECTED}')")
        else:
            print("AUTO_CREATE_MISSING=False: no se crean placeholders de especies.")

# ---------- CREAR TABLA EXACTA CON PK/FKs ----------
metadata = MetaData()

# Reflejar tablas padre en el mismo MetaData para resolver FKs
Table('parcela_inventario', metadata, autoload_with=engine)   # (idparcela, idinventario)
Table(ESPECIES_TABLE, metadata, autoload_with=engine)         # (idespecie)

if not inspector.has_table(TABLE_NAME):
    parcela_inventario_especie_table = Table(
        TABLE_NAME, metadata,
        Column('idparcela',   VARCHAR(60),  nullable=False),
        Column('idinventario', INTEGER(),   nullable=False),
        Column('idespecie',    SMALLINT(),  nullable=False),
        Column('ocupa',    TINYINT(), nullable=True),
        Column('estado',   TINYINT(), nullable=True),
        Column('fpmasa',   TINYINT(), nullable=True),
        Column('tratmasa', TINYINT(), nullable=True),
        Column('orgmasa1', TINYINT(), nullable=True),
        Column('masa',     TINYINT(), nullable=True),
        Column('origen',   TINYINT(), nullable=True),
        PrimaryKeyConstraint('idparcela','idinventario','idespecie', name='pk_pie'),
        ForeignKeyConstraint(
            ['idparcela','idinventario'],
            ['parcela_inventario.idparcela','parcela_inventario.idinventario'],
            name='fk_pie_pi'
        ),
        ForeignKeyConstraint(
            ['idespecie'],
            [f'{ESPECIES_TABLE}.idespecie'],
            name='fk_pie_especie'
        ),
        mysql_charset='utf8mb4', mysql_collate='utf8mb4_unicode_ci'
    )
    metadata.create_all(engine, tables=[parcela_inventario_especie_table])
    print(f"✅ Tabla '{TABLE_NAME}' creada con PK compuesta y FKs.")
else:
    # Hacemos un backup por seguridad; luego insertamos en append.
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    backup_name = f"{TABLE_NAME}_backup_{ts}"
    with engine.begin() as conn:
        conn.execute(text(f"CREATE TABLE `{backup_name}` AS SELECT * FROM `{TABLE_NAME}`;"))
    print(f"ℹ️ Tabla '{TABLE_NAME}' ya existía. Backup creado: {backup_name}. Se insertará en append.")

# ---------- INSERCIÓN ----------
cols_to_insert = [
    'idparcela','idinventario','idespecie',
    'ocupa','estado','fpmasa','tratmasa','orgmasa1','masa','origen'
]
df_ok_to_sql = df_ok[cols_to_insert].copy()

if len(df_ok_to_sql):
    print(f"Insertando {len(df_ok_to_sql):,} filas en '{TABLE_NAME}'…")
    with engine.begin() as conn:
        df_ok_to_sql.to_sql(TABLE_NAME, conn, if_exists='append', index=False, method='multi', chunksize=1000)
    print("✅ Inserción completada.")
else:
    print("No hay filas válidas para insertar.")

# ---------- FIN ----------
```

```python
# -*- coding: utf-8 -*-
from db import engine
import pandas as pd
import numpy as np
import re
from datetime import datetime
from sqlalchemy import (
    create_engine, inspect, text, MetaData, Table, Column,
    ForeignKeyConstraint, PrimaryKeyConstraint
)
from sqlalchemy.dialects.mysql import VARCHAR, TINYINT, SMALLINT, INTEGER, DOUBLE

# ---------------- CONFIG ----------------
ruta_csv = "IFN_España/parcela_inventario_especie_cd.csv"
cols_csv = ['idparcela','inventario','especie','cd','npies','vcc','vsc','vle','iavc','abas','ca','cr']

DO_INSERT = True
CHUNKSIZE = 5000
TABLE_NAME = "parcela_inventario_especie_cd"

# ---------------- Helpers ----------------
def norm_to_int_or_none(x):
    if pd.isna(x) or str(x).strip().lower() in ('', 'nan', 'none', 'null'):
        return None
    s = str(x).strip()
    try:
        return int(float(s))
    except Exception:
        m = re.search(r"-?\d+", s)
        return int(m.group(0)) if m else None

# ---------------- Leer CSV ----------------
print("Leyendo CSV…")
df = pd.read_csv(ruta_csv, usecols=cols_csv, dtype=str)
for c in cols_csv:
    if c not in df.columns:
        df[c] = np.nan

# Normalizar claves
df['idparcela'] = df['idparcela'].astype(str).str.strip()
df['idinventario'] = pd.to_numeric(df['inventario'], errors='coerce').astype('Int64')
df['idespecie'] = df['especie'].apply(norm_to_int_or_none).astype('Int64')
df['cd'] = df['cd'].apply(norm_to_int_or_none).astype('Int64')
df = df.drop(columns=['inventario','especie'])

# Métricas -> DOUBLE
for c in ['npies','vcc','vsc','vle','iavc','abas','ca','cr']:
    df[c] = pd.to_numeric(df[c], errors='coerce')

# Enforce NOT NULL de las claves (aún sin recortar por FK)
required_not_null = ['idparcela','idinventario','idespecie','cd']
before_nn = len(df)
df = df.dropna(subset=required_not_null).copy()
print(f"🧹 Filas eliminadas por claves NULL {required_not_null}: {before_nn - len(df):,}")

# ---------------- Consultar BD (claves padre) ----------------
inspector = inspect(engine)
with engine.connect() as conn:
    # 1) parcelas(idparcela)
    if not inspector.has_table('parcelas'):
        raise SystemExit("No existe la tabla 'parcelas'. Crea 'parcelas' primero.")
    set_parcelas = set(pd.read_sql("SELECT idparcela FROM parcelas", conn)['idparcela'].astype(str).unique())

    # 2) parcela_inventario(idparcela, idinventario)
    if not inspector.has_table('parcela_inventario'):
        raise SystemExit("No existe la tabla 'parcela_inventario'. Crea esa tabla primero.")
    df_pi = pd.read_sql("SELECT idparcela, idinventario FROM parcela_inventario", conn)
    set_pi = set(map(tuple, df_pi.astype({'idparcela':str, 'idinventario':int}).values))

    # 3) especies(idespecie)
    if not inspector.has_table('especies'):
        raise SystemExit("No existe la tabla 'especies'. Crea 'especies' primero.")
    set_especies = set(pd.read_sql("SELECT idespecie FROM especies", conn)['idespecie'].dropna().astype(int).unique())

print(f"Claves en BD -> parcelas: {len(set_parcelas):,} | PI pares: {len(set_pi):,} | especies: {len(set_especies):,}")

# ---------------- Recorte por FKs ----------------
mask_fk_parcela = df['idparcela'].isin(set_parcelas)
mask_fk_pi = df.apply(lambda r: (str(r['idparcela']), int(r['idinventario'])) in set_pi, axis=1)
mask_fk_especie = df['idespecie'].astype(int).isin(set_especies)

keep_mask = mask_fk_parcela & mask_fk_pi & mask_fk_especie
df_ok = df[keep_mask].copy()
df_rej = df[~keep_mask].copy()

print("\n--- Resumen recorte por FK ---")
print(f"Total CSV: {len(df):,}")
print(f"  Pasa FK parcelas(idparcela): {int(mask_fk_parcela.sum()):,}")
print(f"  Pasa FK PI(idparcela,idinventario): {int(mask_fk_pi.sum()):,}")
print(f"  Pasa FK especies(idespecie): {int(mask_fk_especie.sum()):,}")
print(f"✅ Filas que se insertarán (todas las FKs anteriores OK): {len(df_ok):,}")
print(f"❌ Filas rechazadas por alguna FK: {len(df_rej):,}")
# Nota: NO se recorta por (idparcela,idinventario,idespecie) en 'parcela_inventario_especie'

# ---------------- Crear tabla EXACTA (PK+FKs) ----------------
metadata = MetaData()
# Reflejar tablas padre en el mismo MetaData
Table('parcelas', metadata, autoload_with=engine)
Table('parcela_inventario', metadata, autoload_with=engine)
Table('especies', metadata, autoload_with=engine)

if not inspector.has_table(TABLE_NAME):
    piecd = Table(
        TABLE_NAME, metadata,
        Column('idparcela',   VARCHAR(60),  nullable=False),
        Column('idinventario', INTEGER(),   nullable=False),
        Column('idespecie',    SMALLINT(),  nullable=False),
        Column('cd',           INTEGER(),   nullable=False),
        Column('npies', DOUBLE(asdecimal=False), nullable=True),
        Column('vcc',  DOUBLE(asdecimal=False), nullable=True),
        Column('vsc',  DOUBLE(asdecimal=False), nullable=True),
        Column('vle',  DOUBLE(asdecimal=False), nullable=True),
        Column('iavc', DOUBLE(asdecimal=False), nullable=True),
        Column('abas', DOUBLE(asdecimal=False), nullable=True),
        Column('ca',   DOUBLE(asdecimal=False), nullable=True),
        Column('cr',   DOUBLE(asdecimal=False), nullable=True),
        PrimaryKeyConstraint('idparcela','idinventario','idespecie','cd', name='pk_piecd'),
        ForeignKeyConstraint(
            ['idparcela'],
            ['parcelas.idparcela'],
            name='fk_piecd_parcelas'
        ),
        ForeignKeyConstraint(
            ['idparcela','idinventario'],
            ['parcela_inventario.idparcela','parcela_inventario.idinventario'],
            name='fk_piecd_pi'
        ),
        ForeignKeyConstraint(
            ['idespecie'],
            ['especies.idespecie'],
            name='fk_piecd_especies'
        ),
        mysql_charset='utf8mb4', mysql_collate='utf8mb4_unicode_ci'
    )
    metadata.create_all(engine, tables=[piecd])
    print(f"✅ Tabla '{TABLE_NAME}' creada con PK compuesta y FKs.")
else:
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    backup = f"{TABLE_NAME}_backup_{ts}"
    with engine.begin() as conn:
        conn.execute(text(f"CREATE TABLE `{backup}` AS SELECT * FROM `{TABLE_NAME}`;"))
    print(f"ℹ️ La tabla '{TABLE_NAME}' ya existía. Backup creado: {backup}. Insertaremos en append.")

# ---------------- Inserción ----------------
if DO_INSERT:
    print(f"\nInsertando {len(df_ok):,} filas…")
    with engine.begin() as conn:
        for start in range(0, len(df_ok), CHUNKSIZE):
            chunk = df_ok.iloc[start:start+CHUNKSIZE]
            chunk.to_sql(TABLE_NAME, con=conn, if_exists='append', index=False, method='multi', chunksize=CHUNKSIZE)
    print("✅ Inserción completada.")
else:
    print("\nDO_INSERT=False -> sin inserción, solo preparación.")

# ---------------- Resumen final ----------------
print("\n--- RESUMEN FINAL ---")
print(f"Filas preparadas (claves no nulas): {before_nn - (before_nn - len(df)):,}")
print(f"Insertadas (tras recorte por FKs definidas): {len(df_ok):,}")
print(f"Rechazadas por FKs: {len(df_rej):,}")
print("🔎 Recortado por: parcelas(idparcela), parcela_inventario(idparcela,idinventario), especies(idespecie).")
print("🚫 No se recorta por (idparcela,idinventario,idespecie) en 'parcela_inventario_especie'.")
```

```python
# -*- coding: utf-8 -*-
from db import engine
import pandas as pd
import os
from sqlalchemy import text

CSV_IFN2  = "IFN_España/ifn2_seasonal.csv"
CSV_TEST  = "test_descarga/resultado_prueba_100.csv"
CSV_IFN3  = "IFN_España/ifn3_seasonal.csv"
TMP_TABLE = "tmp_ifn_seasonal_insert"
TABLE_WIDE = "parcela_inventario_estacion"

ALL_METRICS = [
    "evi_max","evi_mean","evi_median","evi_min","evi_std",
    "gndvi_max","gndvi_mean","gndvi_median","gndvi_min","gndvi_std",
    "ndii_max","ndii_mean","ndii_median","ndii_min","ndii_std",
    "ndvi_max","ndvi_mean","ndvi_median","ndvi_min","ndvi_std",
    "pr_mean","pr_max","pr_min","pr_std","pr_sum",
    "t2m_mean","t2m_min_mean","t2m_max_mean",
    "skt_mean","skt_min_mean","skt_max_mean",
    "stl1_mean","stl1_min_mean","stl1_max_mean",
    "stl2_mean","stl2_min_mean","stl2_max_mean",
    "stl3_mean","stl3_min_mean","stl3_max_mean",
    "stl4_mean","stl4_min_mean","stl4_max_mean",
    "t2m_min","t2m_min_min","t2m_max_min",
    "skt_min","skt_min_min","skt_max_min",
    "stl1_min","stl1_min_min","stl1_max_min",
    "stl2_min","stl2_min_min","stl2_max_min",
    "stl3_min","stl3_min_min","stl3_max_min",
    "stl4_min","stl4_min_min","stl4_max_min",
    "t2m_max","t2m_min_max","t2m_max_max",
    "skt_max","skt_min_max","skt_max_max",
    "stl1_max","stl1_min_max","stl1_max_max",
    "stl2_max","stl2_min_max","stl2_max_max",
    "stl3_max","stl3_min_max","stl3_max_max",
    "stl4_max","stl4_min_max","stl4_max_max",
    "t2m_std","t2m_min_std","t2m_max_std",
    "skt_std","skt_min_std","skt_max_std",
    "stl1_std","stl1_min_std","stl1_max_std",
    "stl2_std","stl2_min_std","stl2_max_std",
    "stl3_std","stl3_min_std","stl3_max_std",
    "stl4_std","stl4_min_std","stl4_max_std",
]
SEASON_MAP = {
    "spring": "primavera","summer": "verano","fall": "otoño","autumn": "otoño",
    "winter": "invierno","yearly": "yearly"
}

def _run():
    if not os.path.exists(CSV_IFN2) and not os.path.exists(CSV_IFN3) and not os.path.exists(CSV_TEST):
        print("⚠️  Archivos GEE estacionales no encontrados. Celda omitida.")
        return

    def norm_season_es(x):
        if pd.isna(x): return None
        t = str(x).strip().lower()
        return SEASON_MAP.get(t, str(x).strip())

    def load_csv_subset(path):
        df = pd.read_csv(path)
        df = df.rename(columns={"parcela_id": "idparcela", "inventario_id": "inventario", "estacion_id": "season"})
        if "season" not in df.columns:
            raise ValueError(f"Falta columna 'season' en {path}")
        present_metrics = [c for c in ALL_METRICS if c in df.columns]
        keep = ["idparcela","inventario","season"] + present_metrics
        df = df[keep].copy()
        df["idparcela"]  = df["idparcela"].astype(str).str.strip()
        df["inventario"] = pd.to_numeric(df["inventario"], errors="coerce").astype("Int64")
        df["season"]     = df["season"].apply(norm_season_es)
        for c in present_metrics:
            df[c] = pd.to_numeric(df[c], errors="coerce")
        df = df.rename(columns={"season":"estacion"})
        return df, present_metrics

    frames, msets = [], []
    for path in [CSV_IFN2, CSV_IFN3, CSV_TEST]:
        if os.path.exists(path):
            df_i, m_i = load_csv_subset(path)
            frames.append(df_i); msets.append(set(m_i))
        else:
            print(f"⚠️  {path} no encontrado, omitido.")

    metrics = sorted(set().union(*msets))
    cand = pd.concat(frames, ignore_index=True).drop_duplicates(
        subset=["idparcela","inventario","estacion"]
    ).reset_index(drop=True)

    def get_col_collation(conn, table, column):
        q = text("""
            SELECT COLLATION_NAME
            FROM information_schema.COLUMNS
            WHERE TABLE_SCHEMA = DATABASE()
              AND TABLE_NAME = :t AND COLUMN_NAME = :c
        """)
        return conn.execute(q, {"t": table, "c": column}).scalar()

    with engine.begin() as conn:
        collation_idparcela = get_col_collation(conn, TABLE_WIDE, "idparcela") or "utf8mb4_unicode_ci"
        collation_estacion  = get_col_collation(conn, TABLE_WIDE, "estacion")  or "utf8mb4_unicode_ci"
        conn.execute(text(f"DROP TABLE IF EXISTS {TMP_TABLE}"))
        cols_metrics_sql = ", ".join([f"`{c}` DOUBLE NULL" for c in metrics]) if metrics else ""
        conn.execute(text(f"""
            CREATE TABLE {TMP_TABLE} (
                idparcela VARCHAR(60) COLLATE {collation_idparcela} NOT NULL,
                inventario INT NOT NULL,
                estacion  VARCHAR(50) COLLATE {collation_estacion} NOT NULL
                {("," + cols_metrics_sql) if cols_metrics_sql else ""}
               ,PRIMARY KEY (idparcela, inventario, estacion)
            ) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4 COLLATE=utf8mb4_unicode_ci
        """))

    cand.to_sql(TMP_TABLE, engine, if_exists="append", index=False, method="multi", chunksize=5000)

    with engine.begin() as conn:
        coverage_triples = conn.execute(text(f"""
            SELECT COUNT(*) FROM (
              SELECT DISTINCT t.idparcela, t.inventario, t.estacion
              FROM {TMP_TABLE} t
              JOIN parcela_inventario pi
                ON pi.idparcela = t.idparcela COLLATE {collation_idparcela}
               AND pi.idinventario = t.inventario
            ) z
        """)).scalar()
        total_triples = conn.execute(text(f"SELECT COUNT(*) FROM {TMP_TABLE}")).scalar()

    dest_cols   = ["idparcela","idinventario","estacion"] + metrics
    select_cols = ["t.idparcela","t.inventario","t.estacion"] + [f"t.`{c}`" for c in metrics]
    set_updates = ", ".join([f"`{c}`=VALUES(`{c}`)" for c in metrics])

    insert_wide_sql = f"""
INSERT INTO {TABLE_WIDE}
({", ".join(dest_cols)})
SELECT {", ".join(select_cols)}
FROM {TMP_TABLE} t
JOIN parcela_inventario pi
  ON pi.idparcela = t.idparcela COLLATE {collation_idparcela}
 AND pi.idinventario = t.inventario
ON DUPLICATE KEY UPDATE
{set_updates};
"""
    with engine.begin() as conn:
        conn.execute(text(insert_wide_sql))
        total_wide_now = conn.execute(text(f"SELECT COUNT(*) FROM {TABLE_WIDE}")).scalar()
        conn.execute(text(f"DROP TABLE IF EXISTS {TMP_TABLE}"))

    print("=== RESUMEN (WIDE) ===")
    print(f"Triples CSV: {len(cand)}  |  Con cobertura en PI: {coverage_triples} / {total_triples}")
    print(f"Filas totales en {TABLE_WIDE}: {total_wide_now}")

_run()
```

```python
# -*- coding: utf-8 -*-
from db import engine
from sqlalchemy import text
import pandas as pd

TABLE  = "parcela_inventario_estacion"


# Mapeo a español (en minúsculas por si acaso)
MAP = {
    "spring": "primavera",
    "summer": "verano",
    "fall":   "otoño",
    "autumn": "otoño",
    "winter": "invierno",
}

with engine.begin() as conn:
    # 1) Normaliza a minúsculas (opcional pero útil si hay mayúsculas sueltas)
    conn.execute(text(f"UPDATE {TABLE} SET estacion = LOWER(estacion)"))

    # 2) Pasa las etiquetas inglesas a español
    conn.execute(text(f"""
        UPDATE {TABLE}
        SET estacion = CASE
            WHEN estacion = 'spring' THEN 'primavera'
            WHEN estacion = 'summer' THEN 'verano'
            WHEN estacion = 'fall'   THEN 'otoño'
            WHEN estacion = 'autumn' THEN 'otoño'
            WHEN estacion = 'winter' THEN 'invierno'
            ELSE estacion
        END
        WHERE estacion IN ('spring','summer','fall','autumn','winter')
    """))

# 3) Resumen tras la normalización
with engine.connect() as conn:
    df_raw = pd.read_sql_query(
        text(f"SELECT estacion, COUNT(*) AS n FROM {TABLE} GROUP BY estacion ORDER BY n DESC"),
        conn
    )
    print("✅ Normalización hecha. Conteo por estación (ya en español):")
    print(df_raw.to_string(index=False))
```

```python
# -*- coding: utf-8 -*-
from db import engine, DB_NAME
from sqlalchemy import text

SCHEMA = DB_NAME
TABLE   = "parcela_inventario_estacion"


with engine.begin() as conn:
    # 1) Columnas de la PK para excluir
    pk_cols = [r[0] for r in conn.execute(text(f"""
        SELECT k.COLUMN_NAME
        FROM INFORMATION_SCHEMA.TABLE_CONSTRAINTS t
        JOIN INFORMATION_SCHEMA.KEY_COLUMN_USAGE k
          ON k.CONSTRAINT_NAME = t.CONSTRAINT_NAME
         AND k.TABLE_SCHEMA    = t.TABLE_SCHEMA
         AND k.TABLE_NAME      = t.TABLE_NAME
        WHERE t.CONSTRAINT_SCHEMA = :s
          AND t.TABLE_NAME = :t
          AND t.CONSTRAINT_TYPE = 'PRIMARY KEY'
        ORDER BY k.ORDINAL_POSITION
    """), {"s": SCHEMA, "t": TABLE}).fetchall()]

    # 2) Columnas numéricas candidatas (DOUBLE/FLOAT/DECIMAL)
    num_cols = [r[0] for r in conn.execute(text("""
        SELECT COLUMN_NAME
        FROM INFORMATION_SCHEMA.COLUMNS
        WHERE TABLE_SCHEMA = :s
          AND TABLE_NAME   = :t
          AND DATA_TYPE IN ('double','float','decimal')
    """), {"s": SCHEMA, "t": TABLE}).fetchall()]

    # 3) Excluir PK
    targets = [c for c in num_cols if c not in pk_cols]
    if not targets:
        raise SystemExit("No hay columnas numéricas objetivo (o todas son PK).")

    # 4) UPDATE usando NULLIF(col, -999)
    set_clause = ", ".join(f"`{c}` = NULLIF(`{c}`, -999)" for c in targets)
    sql = f"UPDATE `{TABLE}` SET {set_clause}"
    res = conn.execute(text(sql))

    print(f"✅ Hecho. Filas afectadas: {res.rowcount if res.rowcount is not None else 'desconocido'}")
    print(f"Columnas actualizadas ({len(targets)}): {', '.join(targets)}")
```

```python
# -*- coding: utf-8 -*-
from db import engine, DB_NAME
import pandas as pd
from sqlalchemy import text

# ========= CONFIG =========
SCHEMA = DB_NAME   # nombre del esquema/base de datos a documentar
DRY_RUN  = False          # pon True para ver el SQL sin insertar


ddl = f"""
CREATE TABLE IF NOT EXISTS meta_variables (
  id                 INT UNSIGNED NOT NULL AUTO_INCREMENT,
  tabla              VARCHAR(64)  NOT NULL,
  columna            VARCHAR(64)  NOT NULL,
  tipo               VARCHAR(64)  NOT NULL,
  unidades           VARCHAR(64)  NULL,
  descripcion        TEXT         NULL,
  fuente             TEXT         NULL,
  metodo_calculo     TEXT         NULL,
  rango_esperado     VARCHAR(128) NULL,
  valores_especiales VARCHAR(128) NULL,
  PRIMARY KEY (id),
  UNIQUE KEY uk_tabla_col (tabla, columna)
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4 COLLATE=utf8mb4_unicode_ci;
"""

# Inserción esqueleto: todas las columnas del esquema, una fila por columna.
# - Omite la propia tabla meta_variables, por si ya existe en el mismo esquema.
# - Evita duplicados con LEFT JOIN contra meta_variables + WHERE mv.id IS NULL
sql_insert_skeleton = f"""
INSERT INTO meta_variables
  (tabla, columna, tipo, unidades, descripcion, fuente, metodo_calculo, rango_esperado, valores_especiales)
SELECT
  c.TABLE_NAME     AS tabla,
  c.COLUMN_NAME    AS columna,
  c.COLUMN_TYPE    AS tipo,
  NULL             AS unidades,
  NULL             AS descripcion,
  NULL             AS fuente,
  NULL             AS metodo_calculo,
  NULL             AS rango_esperado,
  NULL             AS valores_especiales
FROM information_schema.COLUMNS c
LEFT JOIN {SCHEMA}.meta_variables mv
  ON mv.tabla   = c.TABLE_NAME
 AND mv.columna = c.COLUMN_NAME
WHERE c.TABLE_SCHEMA = :schema
  AND c.TABLE_NAME <> 'meta_variables'
  AND mv.id IS NULL
ORDER BY c.TABLE_NAME, c.ORDINAL_POSITION;
"""

with engine.begin() as conn:
    # 1) Crear tabla si no existe
    conn.execute(text(ddl))

    # 2) Insertar esqueleto evitando duplicados
    if DRY_RUN:
        print("DRY_RUN=True -> no se inserta. SQL preparado:\n", sql_insert_skeleton)
    else:
        result = conn.execute(
            text(sql_insert_skeleton),
            {"schema": SCHEMA}
        )
        print(f"✅ Filas insertadas en meta_variables:", result.rowcount or 0)

# 3) Resumen rápido y muestra de ejemplo
with engine.connect() as conn:
    total = conn.execute(text("SELECT COUNT(*) FROM meta_variables")).scalar()
    print(f"📊 Total filas meta_variables:", int(total))

    # contar por tabla
    df_count = pd.read_sql(text("""
        SELECT tabla, COUNT(*) AS n
        FROM meta_variables
        GROUP BY tabla
        ORDER BY tabla
    """), conn)
    print("🔎 Filas por tabla:")
    print(df_count.head(110).to_string(index=False))

    # sample de 15 filas
    df_sample = pd.read_sql(text("""
        SELECT tabla, columna, tipo, unidades, descripcion
        FROM meta_variables
        ORDER BY tabla, columna
        LIMIT 15
    """), conn)
    print("\n🧪 Muestra de meta_variables:")
    print(df_sample.to_string(index=False))
```

```python
# -*- coding: utf-8 -*-
from db import engine
from sqlalchemy import text
import pandas as pd

TABLA_MD = "meta_variables"
TABLA    = "parcela_inventario_estacion"


# =========================================================
# 1) Asegurar tabla meta_variables (SIN version ni timestamps) + UK(tabla,columna)
# =========================================================
ddl = f"""
CREATE TABLE IF NOT EXISTS {TABLA_MD} (
  id INT AUTO_INCREMENT PRIMARY KEY,
  tabla              VARCHAR(128) NOT NULL,
  columna            VARCHAR(128) NOT NULL,
  tipo               VARCHAR(128) NULL,
  unidades           VARCHAR(128) NULL,
  descripcion        TEXT NULL,
  fuente             TEXT NULL,
  metodo_calculo     TEXT NULL,
  rango_esperado     TEXT NULL,
  valores_especiales TEXT NULL,
  UNIQUE KEY uk_meta_tab_col (tabla, columna)
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4 COLLATE=utf8mb4_unicode_ci;
"""
with engine.begin() as conn:
    conn.execute(text(ddl))

# =========================================================
# 2) Definiciones base (claves, precipitación e índices)
# =========================================================
fuente_era5     = "ERA5-Land (ECMWF), datos horarios 1950–presente."
fuente_indices  = "Índices calculados a partir de reflectancias satelitales (SR_B1/B2/B3/B4/B5)."
val_esp_default = "NA, -9999"

rows = [
    # ---- Claves
    ("idparcela",   "varchar(60)", None,
     "Identificador único de parcela.",
     "Clave externa a tablas de parcelas/IFN.",
     "Agrupación temporal por etiqueta (dic-ene; mar-may, jun-ago, sep-nov)", None, None),

    ("idinventario","int", None,
     "Identificador del inventario IFN.",
     "Asignación por campaña IFN.",
     "Agrupación temporal por etiqueta (dic-ene; mar-may, jun-ago, sep-nov)", "[2,4]", None),

    ("estacion", "varchar(50)", None,
     "Etiqueta del periodo temporal agregado.",
     "Procedimiento interno de agregación temporal.",
     "Días agrupados según calendario estacional.", None, None),

    # ---- Precipitación
    ("pr_mean","double","mm/(m^2 * día)",
     "Precipitación media del periodo.",
     fuente_era5, "Media de acumulados diarios del periodo seleccionado.", ">= 0", val_esp_default),
    ("pr_max","double","mm/(m^2 * día)",
     "Precipitación máxima en el periodo.",
     fuente_era5, "Máximo de los acumulados diarios en el periodo.", ">= 0", val_esp_default),
    ("pr_min","double","mm/(m^2 * día)",
     "Precipitación mínima en el periodo.",
     fuente_era5, "Mínimo de los acumulados diarios en el periodo.", ">= 0", val_esp_default),
    ("pr_std","double","mm/(m^2 * día)",
     "Desviación estándar de la precipitación en el periodo.",
     fuente_era5, "Desviación estándar de los acumulados diarios en el periodo.", ">= 0", val_esp_default),
    ("pr_sum","double","mm/m^2",
     "Precipitación total acumulada en el periodo.",
     fuente_era5, "Suma de los acumulados del periodo.", ">= 0", val_esp_default),
]

# ---- Índices adimensionales
indices = {
    "ndvi": ("Normalized Difference Vegetation Index.",
             "NDVI = (NIR - RED) / (NIR + RED) = (SR_B4 - SR_B3) / (SR_B4 + SR_B3).",
             "[-1, 1]"),
    "ndii": ("Normalized Difference Infrared Index (sensibilidad a contenido de agua foliar).",
             "NDII = (NIR - SWIR) / (NIR + SWIR) = (SR_B4 - SR_B5) / (SR_B4 + SR_B5).",
             "[-1, 1]"),
    "gndvi": ("Green NDVI (sustituye rojo por verde; útil para estrés y contenido de N).",
              "GNDVI = (NIR - GREEN) / (NIR + GREEN) = (SR_B4 - SR_B2) / (SR_B4 + SR_B2).",
              "[-1, 1]"),
    "evi": ("Enhanced Vegetation Index (corrige atmósfera/suelo y reduce saturación).",
            "EVI = 2.5*(SR_B4 - SR_B3)/(SR_B4 + 6*SR_B3 - 7.5*SR_B1 + 1).",
            "[-1, 1.5]"),
}
for idx, (desc, formula, rango) in indices.items():
    for stat in ["max","mean","median","min","std"]:
        rows.append((
            f"{idx}_{stat}", "double", None,
            f"{desc} — {stat}.",
            fuente_indices, formula, rango, val_esp_default
        ))

# =========================================================
# 3) UPSERT para base e índices (sin rangos dinámicos)
# =========================================================
UPSERT = f"""
INSERT INTO {TABLA_MD}
(tabla,columna,tipo,unidades,descripcion,fuente,metodo_calculo,rango_esperado,valores_especiales)
VALUES
(:tabla,:columna,:tipo,:unidades,:descripcion,:fuente,:metodo_calculo,:rango_esperado,:valores_especiales)
ON DUPLICATE KEY UPDATE
  tipo=VALUES(tipo),
  unidades=VALUES(unidades),
  descripcion=VALUES(descripcion),
  fuente=VALUES(fuente),
  metodo_calculo=VALUES(metodo_calculo),
  rango_esperado=VALUES(rango_esperado),
  valores_especiales=VALUES(valores_especiales)
"""
with engine.begin() as conn:
    for (col, tipo, uni, desc, fuente, metodo, rango, valesp) in rows:
        conn.execute(text(UPSERT), {
            "tabla": TABLA,
            "columna": col,
            "tipo": tipo,
            "unidades": uni,
            "descripcion": desc,
            "fuente": fuente,
            "metodo_calculo": metodo,
            "rango_esperado": rango,
            "valores_especiales": valesp
        })

# =========================================================
# 4) Rango dinámico para temperaturas (t2m, skt, stl1..stl4) en °C
#    - Busca columnas existentes por prefijo y calcula min/max reales
# =========================================================
TEMP_VARS = {
    "t2m":  "Temperatura del aire a 2 m sobre superficie terrestre/agua.",
    "skt":  "Temperatura de la superficie terrestre (skin temperature).",
    "stl1": "Temperatura del suelo, nivel 1 (0–7 cm).",
    "stl2": "Temperatura del suelo, nivel 2 (7–28 cm).",
    "stl3": "Temperatura del suelo, nivel 3 (28–100 cm).",
    "stl4": "Temperatura del suelo, nivel 4 (100–289 cm).",
}

def cols_por_prefijo(conn, prefijo):
    q = text("""
        SELECT COLUMN_NAME
        FROM information_schema.COLUMNS
        WHERE TABLE_SCHEMA = DATABASE()
          AND TABLE_NAME   = :tabla
          AND COLUMN_NAME LIKE :likepat
    """)
    rows = conn.execute(q, {"tabla": TABLA, "likepat": f"{prefijo}_%"}).fetchall()
    return sorted([r[0] for r in rows if r[0] not in ("created_at","updated_at")])

def minmax_cols(conn, cols):
    if not cols:
        return None, None
    unions = [f"SELECT {c} AS val FROM {TABLA} WHERE {c} IS NOT NULL" for c in cols]
    sql = "SELECT MIN(val) AS vmin, MAX(val) AS vmax FROM (\n" + "\nUNION ALL\n".join(unions) + "\n) t"
    r = conn.execute(text(sql)).mappings().first()
    return r["vmin"], r["vmax"]

with engine.begin() as conn:
    for pref, desc in TEMP_VARS.items():
        cols = cols_por_prefijo(conn, pref)
        if not cols:
            print(f"⚠️ Sin columnas encontradas para prefijo {pref}")
            continue

        vmin, vmax = minmax_cols(conn, cols)
        if vmin is None or vmax is None:
            rango_txt = None
            print(f"ℹ️ Sólo NULLs para {pref}, no se asigna rango.")
        else:
            # Ya están en °C → NO convertir
            vmin_r = round(float(vmin), 2)
            vmax_r = round(float(vmax), 2)
            rango_txt = f"[{vmin_r}, {vmax_r}] °C"

        for c in cols:
            conn.execute(text(UPSERT), {
                "tabla": TABLA,
                "columna": c,
                "tipo": "double",
                "unidades": "°C",
                "descripcion": desc,
                "fuente": fuente_era5,
                "metodo_calculo": "Agrupar por etiqueta (dic-ene; mar-may, jun-ago, sep-nov)",
                "rango_esperado": rango_txt,
                "valores_especiales": val_esp_default,
            })
        print(f"✅ {pref}: rango {rango_txt} aplicado a {len(cols)} columnas → {cols}")
```

```python
# -*- coding: utf-8 -*-
"""
Carga 'parcela_especie_arbol' desde IFN_España/parcela_especie_arbol.csv
- PK compuesta: (parcela_id, especie_id, arbol_id)
- FK: parcela_id -> parcelas(parcela_id), especie_id -> especies(especie_id)
- Orden de columnas: [parcela_id,especie_id,arbol_id,rumbo,distancia,cd,ht,dn1,dn2,abas,iavc,vcc,vsc,vle,ca,cr]
- Filtra huérfanas (solo inserta si existen parcela y especie en tablas maestras)
"""
from db import engine

import pandas as pd
from sqlalchemy import text
from sqlalchemy.dialects import mysql as mysql_types
from pathlib import Path

# === CONEXIÓN ===
# Requiere: pip install pymysql

# Con mysql-connector, la inserción se hace sin method='multi' y el DDL por separado.

# === RUTA CSV ===
CSV_PATH = Path("IFN_España/parcela_especie_arbol.csv")

# === ESQUEMA DESTINO ===
TABLE_FINAL = "parcela_especie_arbol"
TABLE_STG = "parcela_especie_arbol_stg"

# === ORDEN Y TIPOS (MySQL) ===
COL_ORDER = [
    "parcela_id","especie_id","arbol_id",
    "rumbo","distancia","cd","ht","dn1","dn2",
    "abas","iavc","vcc","vsc","vle","ca","cr"
]

DTYPE_MAP = {
    "parcela_id": mysql_types.VARCHAR(60),
    "especie_id": mysql_types.SMALLINT(unsigned=False),
    "arbol_id":   mysql_types.INTEGER(),
    "rumbo":      mysql_types.DOUBLE(),
    "distancia":  mysql_types.DOUBLE(),
    "cd":         mysql_types.DOUBLE(),
    "ht":         mysql_types.DOUBLE(),
    "dn1":        mysql_types.DOUBLE(),
    "dn2":        mysql_types.DOUBLE(),
    "abas":       mysql_types.DOUBLE(),
    "iavc":       mysql_types.DOUBLE(),
    "vcc":        mysql_types.DOUBLE(),
    "vsc":        mysql_types.DOUBLE(),
    "vle":        mysql_types.DOUBLE(),
    "ca":         mysql_types.DOUBLE(),
    "cr":         mysql_types.DOUBLE(),
}

def main():
    assert CSV_PATH.exists(), f"No se encontró el archivo: {CSV_PATH}"

    # 1) Leer CSV, limpiar y ordenar
    df = pd.read_csv(
        CSV_PATH,
        dtype={
            "parcela_id": "string",
            "especie_id": "Int64",
            "arbol_id":   "Int64",
        }
    )

    if "estrato" in df.columns:
        df = df.drop(columns=["estrato"])

    df = df.rename(columns={"narbol": "arbol_id", "g": "abas"})

    missing = [c for c in COL_ORDER if c not in df.columns]
    if missing:
        raise ValueError(f"Faltan columnas en el CSV: {missing}")

    df = df[COL_ORDER]
    df["parcela_id"] = df["parcela_id"].astype(str).str.strip()

    before = len(df)
    df = df.drop_duplicates(subset=["parcela_id", "especie_id", "arbol_id"], keep="first")
    dedup_removed = before - len(df)

    # 2) Subir a staging (sin 'multi' para evitar resultados pendientes)
    print("Subiendo datos a tabla staging…")
    with engine.begin() as conn:
        df.to_sql(
            TABLE_STG, con=conn, if_exists="replace", index=False,
            dtype=DTYPE_MAP, chunksize=20000  # sin method='multi'
        )

    # 3) Crear tabla final con PK/FK (DDL por separado)
    print("Creando tabla final con PK/FK…")
    drop_sql = f"DROP TABLE IF EXISTS `{TABLE_FINAL}`"
    create_sql = f"""
    CREATE TABLE `{TABLE_FINAL}` (
        `parcela_id` VARCHAR(60)  NOT NULL,
        `especie_id` SMALLINT     NOT NULL,
        `arbol_id`   INT          NOT NULL,
        `rumbo`      DOUBLE NULL,
        `distancia`  DOUBLE NULL,
        `cd`         DOUBLE NULL,
        `ht`         DOUBLE NULL,
        `dn1`        DOUBLE NULL,
        `dn2`        DOUBLE NULL,
        `abas`       DOUBLE NULL,
        `iavc`       DOUBLE NULL,
        `vcc`        DOUBLE NULL,
        `vsc`        DOUBLE NULL,
        `vle`        DOUBLE NULL,
        `ca`         DOUBLE NULL,
        `cr`         DOUBLE NULL,
        PRIMARY KEY (`parcela_id`, `especie_id`, `arbol_id`),
        KEY `idx_pea_parcela` (`parcela_id`),
        KEY `idx_pea_especie` (`especie_id`),
        CONSTRAINT `fk_pea_parcela`
            FOREIGN KEY (`parcela_id`) REFERENCES `parcelas`(`idparcela`)
            ON UPDATE NO ACTION ON DELETE NO ACTION,
        CONSTRAINT `fk_pea_especie`
            FOREIGN KEY (`especie_id`) REFERENCES `especies`(`idespecie`)
            ON UPDATE NO ACTION ON DELETE NO ACTION
    ) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4 COLLATE=utf8mb4_unicode_ci;
    """
    with engine.begin() as conn:
        r1 = conn.execute(text(drop_sql)); r1.close()
        r2 = conn.execute(text(create_sql)); r2.close()

    # 4) Insertar válidas (no huérfanas)
    print("Insertando solo filas válidas (no huérfanas)…")
    insert_sql = f"""
    INSERT INTO `{TABLE_FINAL}` (
        `parcela_id`, `especie_id`, `arbol_id`,
        `rumbo`, `distancia`, `cd`, `ht`, `dn1`, `dn2`,
        `abas`, `iavc`, `vcc`, `vsc`, `vle`, `ca`, `cr`
    )
    SELECT
        s.`parcela_id`, s.`especie_id`, s.`arbol_id`,
        s.`rumbo`, s.`distancia`, s.`cd`, s.`ht`, s.`dn1`, s.`dn2`,
        s.`abas`, s.`iavc`, s.`vcc`, s.`vsc`, s.`vle`, s.`ca`, s.`cr`
    FROM `{TABLE_STG}` s
    INNER JOIN `parcelas`  p ON p.`idparcela`  = s.`parcela_id` COLLATE utf8mb4_unicode_ci
    INNER JOIN `especies`  e ON e.`idespecie`  = s.`especie_id`;
    """
    with engine.begin() as conn:
        res = conn.execute(text(insert_sql))
        inserted = res.rowcount if hasattr(res, "rowcount") else None
        res.close()

    # 5) Métricas y limpieza
    print("Calculando métricas y limpiando staging…")
    with engine.begin() as conn:
        res = conn.execute(text(f"SELECT COUNT(*) FROM `{TABLE_STG}`"))
        total_stg = res.scalar(); res.close()

        res = conn.execute(text(f"""
            SELECT COUNT(*) FROM `{TABLE_STG}` s
            LEFT JOIN `parcelas` p ON p.`idparcela` = s.`parcela_id` COLLATE utf8mb4_unicode_ci
            WHERE p.`idparcela` IS NULL
        """))
        orphan_parcela = res.scalar(); res.close()

        res = conn.execute(text(f"""
            SELECT COUNT(*) FROM `{TABLE_STG}` s
            LEFT JOIN `especies` e ON e.`idespecie` = s.`especie_id`
            WHERE e.`idespecie` IS NULL
        """))
        orphan_especie = res.scalar(); res.close()

        rdrop = conn.execute(text(f"DROP TABLE IF EXISTS `{TABLE_STG}`")); rdrop.close()

    print("\n=== RESUMEN ===")
    print(f"Filas leídas del CSV:                 {before:,}")
    print(f"Filas eliminadas por duplicado PK:    {dedup_removed:,}")
    print(f"Filas en staging:                     {total_stg:,}")
    if inserted is not None:
        print(f"Filas insertadas en final:            {inserted:,}")
    print(f"Huérfanas (parcela):                  {orphan_parcela:,}")
    print(f"Huérfanas (especie):                  {orphan_especie:,}")
    print("\nTabla creada: parcela_especie_arbol ✅")

if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
from db import engine, DB_NAME
from sqlalchemy import text

SCHEMA = DB_NAME
DRY_RUN = False  # True = imprime SQL; False = ejecuta


# --- Utilidades ---
def run(conn, sql, params=None):
    if DRY_RUN:
        print((sql if isinstance(sql, str) else sql.text).strip()+";")
    else:
        conn.execute(text(sql) if isinstance(sql, str) else sql, params or {})

def cols_meta(conn, table, column):
    q = text("""
      SELECT COLUMN_TYPE, IS_NULLABLE, COLUMN_DEFAULT, EXTRA
      FROM INFORMATION_SCHEMA.COLUMNS
      WHERE TABLE_SCHEMA=:s AND TABLE_NAME=:t AND COLUMN_NAME=:c
    """)
    row = conn.execute(q, {"s":SCHEMA,"t":table,"c":column}).first()
    return dict(row._mapping) if row else None

def change_column_sql(table, old, new, meta):
    if not meta: return None
    coltype = meta["COLUMN_TYPE"]            # e.g. 'smallint unsigned'
    nullable = "NULL" if meta["IS_NULLABLE"] == "YES" else "NOT NULL"
    default = meta["COLUMN_DEFAULT"]
    default_sql = ""
    if default is not None:
        # Valor por defecto; manejamos NULL y literales
        if isinstance(default, str):
            default_sql = f" DEFAULT '{default}'"
        else:
            default_sql = f" DEFAULT {default}"
    extra = (" " + meta["EXTRA"]) if meta["EXTRA"] else ""
    return f"ALTER TABLE `{table}` CHANGE COLUMN `{old}` `{new}` {coltype} {nullable}{default_sql}{extra}"

def table_exists(conn, table):
    q = text("""
      SELECT 1 FROM INFORMATION_SCHEMA.TABLES
      WHERE TABLE_SCHEMA=:s AND TABLE_NAME=:t LIMIT 1
    """)
    return conn.execute(q, {"s":SCHEMA,"t":table}).first() is not None

def column_exists(conn, table, col):
    q = text("""
      SELECT 1 FROM INFORMATION_SCHEMA.COLUMNS
      WHERE TABLE_SCHEMA=:s AND TABLE_NAME=:t AND COLUMN_NAME=:c LIMIT 1
    """)
    return conn.execute(q, {"s":SCHEMA,"t":table,"c":col}).first() is not None

def drop_all_fks(conn):
    q = text("""
      SELECT kcu.TABLE_NAME, kcu.CONSTRAINT_NAME
      FROM INFORMATION_SCHEMA.REFERENTIAL_CONSTRAINTS rc
      JOIN INFORMATION_SCHEMA.KEY_COLUMN_USAGE kcu
        ON kcu.CONSTRAINT_SCHEMA = rc.CONSTRAINT_SCHEMA
       AND kcu.CONSTRAINT_NAME   = rc.CONSTRAINT_NAME
       AND kcu.TABLE_NAME        = rc.TABLE_NAME
      WHERE rc.CONSTRAINT_SCHEMA = :s
      GROUP BY kcu.TABLE_NAME, kcu.CONSTRAINT_NAME
    """)
    rows = conn.execute(q, {"s":SCHEMA}).fetchall()
    for t, cname in rows:
        run(conn, f"ALTER TABLE `{t}` DROP FOREIGN KEY `{cname}`")

# --- Mapeos a renombrar (solo PKs/compuestas pedidas) ---
# Catálogos: cat_X.id -> X_id (X = nombre tras 'cat_')
CAT_TABLES = [
    "cat_comesp","cat_disesp","cat_erosion","cat_estado","cat_fpmasa","cat_masa",
    "cat_matorg","cat_merosiva","cat_modcomb","cat_nivel1","cat_nivel2","cat_orgmasa1",
    "cat_origen","cat_presencia","cat_rocosidad","cat_textura","cat_tipsuelo1",
    "cat_tipsuelo2","cat_tipsuelo3","cat_tratmasa"
]

# Tablas y columnas a renombrar explícitamente (claves primarias)
RENAME_SPEC = {
    # Maestras
    "grupos": {"idgrupo": "grupo_id"},
    "especies": {"idespecie": "especie_id", "grupo": "grupo_id"},

    # Parcelas y derivados
    "parcelas": {"idparcela": "parcela_id"},

    "parcela_inventario": {
        "idparcela": "parcela_id",
        "idinventario": "inventario_id",
    },

    "parcela_inventario_especie": {
        "idparcela": "parcela_id",
        "idinventario": "inventario_id",
        "idespecie": "especie_id",
    },

    "parcela_inventario_especie_cd": {
        "idparcela": "parcela_id",
        "idinventario": "inventario_id",
        "idespecie": "especie_id",
        "cd": "cd_id",
    },

    "parcela_inventario_estacion": {
        "idparcela": "parcela_id",
        "idinventario": "inventario_id",
        "estacion": "estacion_id",
    },
}

with engine.begin() as conn:
    # 0) Quitar FKs para evitar bloqueos
    run(conn, "SET FOREIGN_KEY_CHECKS=0")
    drop_all_fks(conn)

    # 1) Catálogos: id -> <var>_id
    for t in CAT_TABLES:
        if not table_exists(conn, t): 
            continue
        # nombre base = después de 'cat_'
        base = t[4:]
        new_pk = f"{base}_id"
        if column_exists(conn, t, "id") and not column_exists(conn, t, new_pk):
            meta = cols_meta(conn, t, "id")
            sql = change_column_sql(t, "id", new_pk, meta)
            if sql: run(conn, sql)

    # 2) Resto de PKs pedidas
    for table, renames in RENAME_SPEC.items():
        if not table_exists(conn, table):
            continue
        for old, new in renames.items():
            if column_exists(conn, table, old) and not column_exists(conn, table, new):
                meta = cols_meta(conn, table, old)
                sql = change_column_sql(table, old, new, meta)
                if sql: run(conn, sql)

    # 3) Volver a activar chequeo FK (sin recrearlas todavía)
    run(conn, "SET FOREIGN_KEY_CHECKS=1")

print("✅ FKs eliminadas y PKs renombradas según la convención *_id. (FKs se declararán en la siguiente celda)")
```

```python
# -*- coding: utf-8 -*-
from db import engine, DB_NAME
from sqlalchemy import text

SCHEMA = DB_NAME
DRY_RUN = False  # True = solo imprime; False = ejecuta


def run(conn, sql, params=None):
    if DRY_RUN:
        print((sql if isinstance(sql, str) else sql.text).strip()+";")
    else:
        conn.execute(text(sql) if isinstance(sql, str) else sql, params or {})

def table_exists(conn, t):
    q = text("""SELECT 1 FROM information_schema.tables
                WHERE table_schema=:s AND table_name=:t""")
    return conn.execute(q, {"s":SCHEMA,"t":t}).first() is not None

def col_exists(conn, t, c):
    q = text("""SELECT 1 FROM information_schema.columns
                WHERE table_schema=:s AND table_name=:t AND column_name=:c""")
    return conn.execute(q, {"s":SCHEMA,"t":t,"c":c}).first() is not None

def col_meta(conn, t, c):
    q = text("""SELECT COLUMN_TYPE, IS_NULLABLE, COLUMN_DEFAULT, EXTRA
                FROM information_schema.columns
                WHERE table_schema=:s AND table_name=:t AND column_name=:c""")
    r = conn.execute(q, {"s":SCHEMA,"t":t,"c":c}).first()
    return dict(r._mapping) if r else None

def change_sql(t, old, new, meta):
    if not meta: return None
    coltype  = meta["COLUMN_TYPE"]
    nullable = "NULL" if meta["IS_NULLABLE"]=="YES" else "NOT NULL"
    default  = meta["COLUMN_DEFAULT"]
    default_sql = ""
    if default is not None:
        default_sql = f" DEFAULT {default!r}" if isinstance(default, str) else f" DEFAULT {default}"
    extra = f" {meta['EXTRA']}" if meta["EXTRA"] else ""
    return f"ALTER TABLE `{t}` CHANGE COLUMN `{old}` `{new}` {coltype} {nullable}{default_sql}{extra}"

def drop_fk(conn, table, name):
    run(conn, f"ALTER TABLE `{table}` DROP FOREIGN KEY `{name}`")

def existing_fks(conn):
    q = text("""
      SELECT kcu.TABLE_NAME, kcu.CONSTRAINT_NAME
      FROM information_schema.referential_constraints rc
      JOIN information_schema.key_column_usage kcu
        ON kcu.constraint_schema=rc.constraint_schema
       AND kcu.constraint_name=rc.constraint_name
       AND kcu.table_name=rc.table_name
      WHERE rc.constraint_schema=:s
      GROUP BY kcu.TABLE_NAME, kcu.CONSTRAINT_NAME
    """)
    return [(r[0], r[1]) for r in conn.execute(q, {"s":SCHEMA}).fetchall()]

def orphan_count(conn, child_tbl, child_cols, parent_tbl, parent_cols, where_not_null=True):
    # LEFT JOIN y cuenta hijos sin padre (si where_not_null=True, ignora hijos con NULL en la FK)
    on = " AND ".join([f"c.`{cc}` = p.`{pc}`" for cc, pc in zip(child_cols, parent_cols)])
    nn = " AND ".join([f"c.`{cc}` IS NOT NULL" for cc in child_cols]) if where_not_null else "1=1"
    sql = f"""
        SELECT COUNT(*) FROM `{child_tbl}` c
        LEFT JOIN `{parent_tbl}` p ON {on}
        WHERE {nn} AND p.`{parent_cols[0]}` IS NULL
    """
    return conn.execute(text(sql)).scalar()

def add_fk(conn, child_tbl, child_cols, parent_tbl, parent_cols, name=None,
           on_delete="RESTRICT", on_update="RESTRICT"):
    cname = name or f"fk_{child_tbl}__{'_'.join(child_cols)}__{parent_tbl}"
    cols_sql = ", ".join(f"`{c}`" for c in child_cols)
    ref_sql  = ", ".join(f"`{c}`" for c in parent_cols)
    clause = ""
    if on_delete and on_delete.upper()!="RESTRICT": clause += f" ON DELETE {on_delete}"
    if on_update and on_update.upper()!="RESTRICT": clause += f" ON UPDATE {on_update}"
    run(conn, f"ALTER TABLE `{child_tbl}` ADD CONSTRAINT `{cname}` "
              f"FOREIGN KEY ({cols_sql}) REFERENCES `{parent_tbl}` ({ref_sql}){clause}")

# --------- 1) Renombrar columnas FK a *_id ---------
RENAME = {
    # parcelas
    ("parcelas", "presencia"):  "presencia_id",
    ("parcelas", "rocosidad"):  "rocosidad_id",
    ("parcelas", "tipsuelo1"):  "tipsuelo1_id",
    ("parcelas", "tipsuelo2"):  "tipsuelo2_id",
    ("parcelas", "tipsuelo3"):  "tipsuelo3_id",

    # parcela_inventario
    ("parcela_inventario","nivel1"):   "nivel1_id",
    ("parcela_inventario","nivel2"):   "nivel2_id",
    ("parcela_inventario","merosiva"): "merosiva_id",
    ("parcela_inventario","textura"):  "textura_id",
    ("parcela_inventario","matorg"):   "matorg_id",
    ("parcela_inventario","modcomb"):  "modcomb_id",
    ("parcela_inventario","dise sp".replace(" ","")): "dise sp_id".replace(" ",""),  # safety if typo
    ("parcela_inventario","disesp"):   "disesp_id",
    ("parcela_inventario","comesp"):   "comesp_id",

    # parcela_inventario_especie
    ("parcela_inventario_especie","estado"):   "estado_id",
    ("parcela_inventario_especie","fpmasa"):   "fpmasa_id",
    ("parcela_inventario_especie","tratmasa"): "tratmasa_id",
    ("parcela_inventario_especie","orgmasa1"): "orgmasa1_id",
    ("parcela_inventario_especie","masa"):     "masa_id",
    ("parcela_inventario_especie","origen"):   "origen_id",
}

created, skipped = [], []
with engine.begin() as conn:
    run(conn, "SET FOREIGN_KEY_CHECKS=0")
    # por si aún quedara alguna FK suelta
    for t, cname in existing_fks(conn):
        drop_fk(conn, t, cname)

    for (t, old), new in RENAME.items():
        if not table_exists(conn, t): 
            skipped.append((t, old, "tabla no existe"))
            continue
        if col_exists(conn, t, new):
            skipped.append((t, old, "ya existe destino"))
            continue
        if not col_exists(conn, t, old):
            skipped.append((t, old, "no existe origen"))
            continue
        meta = col_meta(conn, t, old)
        sql  = change_sql(t, old, new, meta)
        if sql:
            run(conn, sql); created.append((t, old, new))
    run(conn, "SET FOREIGN_KEY_CHECKS=1")

print(f"🔧 Renombradas {len(created)} columnas a *_id. Saltadas: {len(skipped)}")
```

```python
# -*- coding: utf-8 -*-
from db import engine, DB_NAME
from sqlalchemy import text

SCHEMA = DB_NAME
DRY_RUN = False  # True = solo imprime; False = ejecuta


TARGETS = [
    # (tabla_hija, columna_fk, tabla_catalogo, pk_catalogo)
    ("parcela_inventario_especie", "origen_id",  "cat_origen",  "origen_id"),
    ("parcela_inventario",         "disesp_id",  "cat_disesp",  "disesp_id"),
    ("parcela_inventario",         "comesp_id",  "cat_comesp",  "comesp_id"),
    ("parcela_inventario_especie", "orgmasa1_id","cat_orgmasa1","orgmasa1_id"),
]

def count_orphans(conn, child, col, cat, cat_pk):
    sql = f"""
        SELECT COUNT(*) 
        FROM `{child}` c
        LEFT JOIN `{cat}` k ON c.`{col}` = k.`{cat_pk}`
        WHERE c.`{col}` IS NOT NULL AND k.`{cat_pk}` IS NULL
    """
    return conn.execute(text(sql)).scalar()

def nullify_orphans(conn, child, col, cat, cat_pk):
    sql = f"""
        UPDATE `{child}` c
        LEFT JOIN `{cat}` k ON c.`{col}` = k.`{cat_pk}`
        SET c.`{col}` = NULL
        WHERE c.`{col}` IS NOT NULL AND k.`{cat_pk}` IS NULL
    """
    if DRY_RUN:
        print(sql.strip()+";")
        return None
    res = conn.execute(text(sql))
    try:
        return res.rowcount
    except Exception:
        return None

before = []
after  = []
changed = []

with engine.begin() as conn:
    # 1) Conteo previo
    for child, col, cat, cat_pk in TARGETS:
        n = count_orphans(conn, child, col, cat, cat_pk)
        before.append((child, col, cat, cat_pk, n))

    # 2) Nullificar huérfanas
    for child, col, cat, cat_pk in TARGETS:
        rc = nullify_orphans(conn, child, col, cat, cat_pk)
        changed.append((child, col, rc))

    # 3) Conteo posterior
    for child, col, cat, cat_pk in TARGETS:
        n = count_orphans(conn, child, col, cat, cat_pk)
        after.append((child, col, cat, cat_pk, n))

print("=== HUÉRFANAS (antes) ===")
for child, col, cat, cat_pk, n in before:
    print(f"- {child}({col}) → {cat}({cat_pk}): {n}")

print("\n=== FILAS MODIFICADAS ===")
for child, col, rc in changed:
    print(f"- {child}.{col}: {rc if rc is not None else '(DRY_RUN)'}")

print("\n=== HUÉRFANAS (después) ===")
for child, col, cat, cat_pk, n in after:
    print(f"- {child}({col}) → {cat}({cat_pk}): {n}")
```

```python
# -*- coding: utf-8 -*-
from db import engine, DB_NAME
from sqlalchemy import text
import hashlib

SCHEMA = DB_NAME
DRY_RUN = False  # True = solo imprime; False = ejecuta


def run(conn, sql, params=None):
    if DRY_RUN:
        print((sql if isinstance(sql, str) else sql.text).strip()+";")
    else:
        conn.execute(text(sql) if isinstance(sql, str) else sql, params or {})

def table_exists(conn, t):
    q = text("""SELECT 1 FROM information_schema.tables
                WHERE table_schema=:s AND table_name=:t""")
    return conn.execute(q, {"s":SCHEMA,"t":t}).first() is not None

def col_exists(conn, t, c):
    q = text("""SELECT 1 FROM information_schema.columns
                WHERE table_schema=:s AND table_name=:t AND column_name=:c""")
    return conn.execute(q, {"s":SCHEMA,"t":t,"c":c}).first() is not None

def fk_abbr(name: str) -> str:
    # abreviatura: primeras letras de cada token (máx 6)
    parts = [p for p in name.split("_") if p]
    ab = "".join(p[0] for p in parts)[:6]
    return ab or name[:6]

def fk_hash(child, ccols, parent, pcols) -> str:
    s = f"{child}|{','.join(ccols)}|{parent}|{','.join(pcols)}"
    return hashlib.md5(s.encode("utf-8")).hexdigest()[:8]

def make_fk_name(child, ccols, parent, pcols) -> str:
    return f"fk_{fk_abbr(child)}_{fk_abbr(parent)}_{fk_hash(child, ccols, parent, pcols)}"

def orphan_count(conn, child_tbl, child_cols, parent_tbl, parent_cols, where_not_null=True):
    on = " AND ".join([f"c.`{cc}` = p.`{pc}`" for cc, pc in zip(child_cols, parent_cols)])
    nn = " AND ".join([f"c.`{cc}` IS NOT NULL" for cc in child_cols]) if where_not_null else "1=1"
    sql = f"""
        SELECT COUNT(*) FROM `{child_tbl}` c
        LEFT JOIN `{parent_tbl}` p ON {on}
        WHERE {nn} AND p.`{parent_cols[0]}` IS NULL
    """
    return conn.execute(text(sql)).scalar()

def fk_already_exists(conn, child_tbl, child_cols, parent_tbl, parent_cols) -> bool:
    # ¿Existe ya una FK con misma tabla hija, mismas columnas (en orden) y mismo padre?
    q = text("""
      SELECT rc.CONSTRAINT_NAME
      FROM information_schema.referential_constraints rc
      JOIN information_schema.key_column_usage kcu
        ON kcu.constraint_schema=rc.constraint_schema
       AND kcu.constraint_name=rc.constraint_name
       AND kcu.table_name=rc.table_name
      WHERE rc.constraint_schema=:s AND kcu.table_name=:t
      GROUP BY rc.CONSTRAINT_NAME
    """)
    rows = conn.execute(q, {"s":SCHEMA, "t": child_tbl}).fetchall()
    for (cname,) in rows:
        q2 = text("""
          SELECT kcu.COLUMN_NAME, kcu.REFERENCED_TABLE_NAME, kcu.REFERENCED_COLUMN_NAME
          FROM information_schema.key_column_usage kcu
          WHERE kcu.constraint_schema=:s
            AND kcu.table_name=:t
            AND kcu.constraint_name=:c
          ORDER BY kcu.ORDINAL_POSITION
        """)
        cols = conn.execute(q2, {"s":SCHEMA, "t":child_tbl, "c":cname}).fetchall()
        fk_cols  = [r[0] for r in cols]
        ref_tbls = list({r[1] for r in cols})
        ref_cols = [r[2] for r in cols]
        if fk_cols == child_cols and len(ref_tbls)==1 and ref_tbls[0]==parent_tbl and ref_cols == parent_cols:
            return True
    return False

def add_fk(conn, child_tbl, child_cols, parent_tbl, parent_cols, on_delete="RESTRICT", on_update="RESTRICT"):
    cname = make_fk_name(child_tbl, child_cols, parent_tbl, parent_cols)
    # MySQL límite 64 chars
    if len(cname) > 64:
        cname = cname[:64]
    cols_sql = ", ".join(f"`{c}`" for c in child_cols)
    ref_sql  = ", ".join(f"`{c}`" for c in parent_cols)
    clause = ""
    if on_delete and on_delete.upper()!="RESTRICT": clause += f" ON DELETE {on_delete}"
    if on_update and on_update.upper()!="RESTRICT": clause += f" ON UPDATE {on_update}"
    run(conn, f"ALTER TABLE `{child_tbl}` ADD CONSTRAINT `{cname}` "
              f"FOREIGN KEY ({cols_sql}) REFERENCES `{parent_tbl}` ({ref_sql}){clause}")

# Reglas FK a crear
FK_RULES = [
    # Jerarquía parcela/inventario/especie
    ("parcela_inventario",          ["parcela_id"],                 "parcelas",                   ["parcela_id"]),
    ("parcela_inventario_especie",  ["parcela_id","inventario_id"], "parcela_inventario",         ["parcela_id","inventario_id"]),
    ("parcela_inventario_especie_cd",["parcela_id","inventario_id","especie_id"],
                                     "parcela_inventario_especie",  ["parcela_id","inventario_id","especie_id"]),
    ("parcela_inventario_estacion", ["parcela_id","inventario_id"], "parcela_inventario",         ["parcela_id","inventario_id"]),

    # especie / grupo
    ("especies", ["grupo_id"], "grupos", ["grupo_id"]),
    ("parcela_inventario_especie", ["especie_id"], "especies", ["especie_id"]),
    ("parcela_inventario_especie_cd", ["especie_id"], "especies", ["especie_id"]),

    # catálogos en parcelas
    ("parcelas", ["presencia_id"], "cat_presencia", ["presencia_id"]),
    ("parcelas", ["rocosidad_id"], "cat_rocosidad", ["rocosidad_id"]),
    ("parcelas", ["tipsuelo1_id"], "cat_tipsuelo1", ["tipsuelo1_id"]),
    ("parcelas", ["tipsuelo2_id"], "cat_tipsuelo2", ["tipsuelo2_id"]),
    ("parcelas", ["tipsuelo3_id"], "cat_tipsuelo3", ["tipsuelo3_id"]),

    # catálogos en parcela_inventario
    ("parcela_inventario", ["nivel1_id"],  "cat_nivel1",  ["nivel1_id"]),
    ("parcela_inventario", ["nivel2_id"],  "cat_nivel2",  ["nivel2_id"]),
    ("parcela_inventario", ["merosiva_id"],"cat_merosiva",["merosiva_id"]),
    ("parcela_inventario", ["textura_id"], "cat_textura", ["textura_id"]),
    ("parcela_inventario", ["matorg_id"],  "cat_matorg",  ["matorg_id"]),
    ("parcela_inventario", ["modcomb_id"], "cat_modcomb", ["modcomb_id"]),
    ("parcela_inventario", ["disesp_id"],  "cat_disesp",  ["disesp_id"]),
    ("parcela_inventario", ["comesp_id"],  "cat_comesp",  ["comesp_id"]),

    # catálogos en parcela_inventario_especie
    ("parcela_inventario_especie", ["estado_id"],   "cat_estado",   ["estado_id"]),
    ("parcela_inventario_especie", ["fpmasa_id"],   "cat_fpmasa",   ["fpmasa_id"]),
    ("parcela_inventario_especie", ["tratmasa_id"], "cat_tratmasa", ["tratmasa_id"]),
    ("parcela_inventario_especie", ["orgmasa1_id"], "cat_orgmasa1", ["orgmasa1_id"]),
    ("parcela_inventario_especie", ["masa_id"],     "cat_masa",     ["masa_id"]),
    ("parcela_inventario_especie", ["origen_id"],   "cat_origen",   ["origen_id"]),
]

report = []
created_fk = 0

with engine.begin() as conn:
    # No soltamos todas las FKs: reintentamos solo las que faltan
    for child, ccols, parent, pcols in FK_RULES:
        # Validaciones básicas
        if not table_exists(conn, child) or not table_exists(conn, parent):
            report.append((child, ccols, parent, pcols, "SKIP: tabla no existe", None))
            continue
        if any(not col_exists(conn, child, c) for c in ccols) or any(not col_exists(conn, parent, p) for p in pcols):
            report.append((child, ccols, parent, pcols, "SKIP: columna no existe", None))
            continue

        # ¿Ya existe una FK equivalente?
        if fk_already_exists(conn, child, ccols, parent, pcols):
            report.append((child, ccols, parent, pcols, "OK (ya existe)", 0))
            continue

        # Huérfanas
        n_orph = orphan_count(conn, child, ccols, parent, pcols, where_not_null=True)
        if n_orph > 0:
            report.append((child, ccols, parent, pcols, "HUERFANAS", n_orph))
            continue

        # Crear con nombre corto determinista
        try:
            add_fk(conn, child, ccols, parent, pcols, on_delete="RESTRICT", on_update="RESTRICT")
            created_fk += 1
            report.append((child, ccols, parent, pcols, "OK (creada)", 0))
        except Exception as e:
            report.append((child, ccols, parent, pcols, f"ERROR: {e.__class__.__name__}", None))

# Informe
print("=== RESUMEN FKs (nombres cortos) ===")
print(f"FKs creadas ahora: {created_fk} / {len(FK_RULES)}")
for child, ccols, parent, pcols, status, n in report:
    cols = ",".join(ccols); pks = ",".join(pcols)
    msg = f"{child}({cols}) → {parent}({pks}) : {status}"
    if n is not None: msg += f" = {n}"
    print(" -", msg)

print("\n⚠️ Donde aparece 'HUERFANAS' no se creó la FK. Corrige esos valores (o ponlos a NULL) y vuelve a ejecutar.")
```

```python
import os
if not os.path.exists("meta_variables2.csv"):
    print("⚠️  meta_variables2.csv no encontrado. Celda omitida.")
else:
    from db import engine
    import pandas as pd
    from sqlalchemy import text
    
    # --- Conexión ---
    
    # --- Carga CSV ---
    csv_path = "meta_variables2.csv"
    df = pd.read_csv(csv_path, dtype=str, keep_default_na=False, encoding="utf-8")
    
    # Ignorar 'id' del CSV y normalizar nombres
    cols_map = {
        "Norma o referencia": "norma"
    }
    df = df.drop(columns=[c for c in ["id"] if c in df.columns]).rename(columns=cols_map)
    
    # Asegurar presencia de todas las columnas esperadas del destino temporal
    expected_cols = [
        "tabla","columna","tipo","unidades","descripcion","fuente",
        "metodo_calculo","rango_esperado","valores_especiales","norma"
    ]
    for c in expected_cols:
        if c not in df.columns:
            df[c] = ""
    
    # (Opcional) recortar espacios
    df = df[expected_cols].applymap(lambda x: x.strip() if isinstance(x, str) else x)
    
    tmp_table = "tmp_meta_variables"
    
    with engine.begin() as conn:
        # 1) Crear/llenar tabla temporal con los datos del CSV
        #    REPLACE para dejar solo los datos del CSV en la tabla temporal
        df.to_sql(tmp_table, conn, if_exists="replace", index=False)
    
        # 2) Añadir columna 'norma' en meta_variables (si no existiera)
        try:
            conn.execute(text("""
                ALTER TABLE meta_variables
                ADD COLUMN norma TEXT NULL
            """))
        except Exception as e:
            # Si ya existe, MySQL lanzará error; lo ignoramos de forma segura
            pass
    
        # 3) Actualizar meta_variables con JOIN por (tabla, columna)
        #    Solo pisa si el CSV trae valor no vacío
        update_sql = text("""
            UPDATE meta_variables mv
            JOIN tmp_meta_variables t
              ON mv.tabla = t.tabla AND mv.columna = t.columna
            SET
              mv.tipo              = COALESCE(NULLIF(t.tipo, ''), mv.tipo),
              mv.unidades          = COALESCE(NULLIF(t.unidades, ''), mv.unidades),
              mv.descripcion       = COALESCE(NULLIF(t.descripcion, ''), mv.descripcion),
              mv.fuente            = COALESCE(NULLIF(t.fuente, ''), mv.fuente),
              mv.metodo_calculo    = COALESCE(NULLIF(t.metodo_calculo, ''), mv.metodo_calculo),
              mv.rango_esperado    = COALESCE(NULLIF(t.rango_esperado, ''), mv.rango_esperado),
              mv.valores_especiales= COALESCE(NULLIF(t.valores_especiales, ''), mv.valores_especiales),
              mv.norma             = COALESCE(NULLIF(t.norma, ''), mv.norma)
        """)
        conn.execute(update_sql)
    
        # (Opcional) comprobar cuántas filas coinciden y se han podido actualizar
        res = conn.execute(text("""
            SELECT COUNT(*) AS n_match
            FROM meta_variables mv
            JOIN tmp_meta_variables t
              ON mv.tabla=t.tabla AND mv.columna=t.columna
        """)).mappings().one()
        print(f"Filas coincidentes (tabla+columna): {res['n_match']}")
    
    print("✅ Actualización completada.")
```
